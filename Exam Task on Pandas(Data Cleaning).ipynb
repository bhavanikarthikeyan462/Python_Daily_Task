```python
# Q30 — Ultimate Pandas Cleaning Project

import pandas as pd

# step 1:
df = pd.read_csv("C:\\Users\\91965\\OneDrive\\Desktop\\.ipynb_checkpoints\\Data Science Course\\customer_analytics.csv")
```

```python
# STEP 2 — INSPECT

print(df.head(10))

print(df.tail(10))

print(df.sample(10))

df.info()
```

```python
# STEP 3 — MISSING VALUE ANALYSIS

print(df.isnull().sum())

print(df.isna().sum())
```

```python
# STEP 4 — DUPLICATES

print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after cleaning:",
      df.duplicated().sum())
```

```python
# STEP 5 — NUMERIC CLEANING

# Age
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"] = df["Age"].fillna(df["Age"].median())


# Annual Income
df["Annual_Income"] = pd.to_numeric(
    df["Annual_Income"],
    errors="coerce"
)

df["Annual_Income"] = df["Annual_Income"].fillna(
    df["Annual_Income"].median()
)

df["Annual_Income"] = df["Annual_Income"].astype(float)


# Account Balance
df["Account_Balance"] = pd.to_numeric(
    df["Account_Balance"],
    errors="coerce"
)

df["Account_Balance"] = df["Account_Balance"].fillna(
    df["Account_Balance"].mean()
)


# Credit Score
df["Credit_Score"] = pd.to_numeric(
    df["Credit_Score"],
    errors="coerce"
)

df["Credit_Score"] = df["Credit_Score"].fillna(
    df["Credit_Score"].median()
)

df["Credit_Score"] = df["Credit_Score"].astype(int)


# Monthly Spend
df["Monthly_Spend"] = pd.to_numeric(
    df["Monthly_Spend"],
    errors="coerce"
)

df["Monthly_Spend"] = df["Monthly_Spend"].fillna(
    df["Monthly_Spend"].median()
)


# Transactions
df["Transactions"] = pd.to_numeric(
    df["Transactions"],
    errors="coerce"
)

df["Transactions"] = df["Transactions"].fillna(
    df["Transactions"].median()
)

df["Transactions"] = df["Transactions"].astype(int)


# Satisfaction Score
df["Satisfaction_Score"] = pd.to_numeric(
    df["Satisfaction_Score"],
    errors="coerce"
)

df["Satisfaction_Score"] = df["Satisfaction_Score"].fillna(
    df["Satisfaction_Score"].mean()
)

df["Satisfaction_Score"] = df[
    "Satisfaction_Score"
].astype(float)
df["Age"]
df["Annual_Income"]
df["Account_Balance"]
df["Credit_Score"]
df["Monthly_Spend"]
df["Transactions"]
df["Satisfaction_Score"]
```

```python
# STEP 6 — DATE CLEANING

df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    errors="coerce"
)

valid_dates = df["Join_Date"].dropna().sort_values()

median_date = valid_dates.iloc[len(valid_dates) // 2]

df["Join_Date"] = df["Join_Date"].fillna(median_date)
df["Join_Date"]
```

```python
# STEP 7 — CATEGORICAL CLEANING

# City
df["City"] = df["City"].replace({
    "mumbai": "Mumbai",
    "MUMBAI": "Mumbai",
    "delhi": "Delhi",
    "DELHI": "Delhi",
    "bangalore": "Bangalore",
    "BANGALORE": "Bangalore",
    "hyderabad": "Hyderabad",
    "HYDERABAD": "Hyderabad",
    "chennai": "Chennai",
    "CHENNAI": "Chennai"
})


# Segment
df["Segment"] = df["Segment"].replace({
    "retail": "Retail",
    "RETAIL": "Retail",
    "sme": "SME",
    "SME": "SME",
    "corporate": "Corporate",
    "CORPORATE": "Corporate",
    "premium": "Premium",
    "PREMIUM": "Premium"
})

df["Segment"] = df["Segment"].fillna(
    df["Segment"].value_counts().index[0]
)


# Channel
df["Channel"] = df["Channel"].replace({
    "online": "Online",
    "ONLINE": "Online",
    "branch": "Branch",
    "BRANCH": "Branch",
    "mobile": "Mobile",
    "MOBILE": "Mobile"
})

df["Channel"] = df["Channel"].fillna(
    df["Channel"].value_counts().index[0]
)


# Active
df["Active"] = df["Active"].replace({
    "Y": "Yes",
    "YES": "Yes",
    "yes": "Yes",
    "1": "Yes",
    1: "Yes",

    "N": "No",
    "NO": "No",
    "no": "No",
    "0": "No",
    0: "No"
})

df["Active"] = df["Active"].fillna(
    df["Active"].value_counts().index[0]
)


# Risk Level
df["Risk_Level"] = df["Risk_Level"].replace({
    "high": "High",
    "HIGH": "High",
    "medium": "Medium",
    "MEDIUM": "Medium",
    "low": "Low",
    "LOW": "Low"
})

df["Risk_Level"] = df["Risk_Level"].fillna(
    df["Risk_Level"].value_counts().index[0]
)
df["Risk_Level"] 
```

```python
# STEP 8 — COLUMN TRANSFORMATION

df = df.rename(columns={
    "Customer_ID": "Customer_ID",
    "Annual_Income": "AnnualIncome",
    "Account_Balance": "AccountBalance",
    "Credit_Score": "CreditScore",
    "Tenure_Years": "Tenure",
    "Join_Date": "CustomerJoinDate",
    "Risk_Level": "RiskLevel",
    "Monthly_Spend": "MonthlySpend",
    "Satisfaction_Score": "SatisfactionScore"
})
df
```

```python
# STEP 9 — FINAL VALIDATION

print("\n========== FINAL VALIDATION ==========")

# No duplicate rows
print("Duplicate rows:")
print(df.duplicated().sum())


# Missing values
print("\nMissing values:")
print(df.isnull().sum())


# Datatypes
print("\nDatatypes:")
print(df.dtypes)


# Numeric columns
print("\nNumeric columns:")
print(df.select_dtypes(include="number").columns)


# City distribution
print("\nCity distribution:")
print(df["Customer_City"].value_counts())


# Segment distribution
print("\nSegment distribution:")
print(df["Customer_Segment"].value_counts())


# Channel distribution
print("\nChannel distribution:")
print(df["Customer_Channel"].value_counts())


# Risk distribution
print("\nRisk Level distribution:")
print(df["Customer_RiskLevel"].value_counts())


```
