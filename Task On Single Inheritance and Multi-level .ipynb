```python
# 6. BankAccount → SavingsAccount

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")


class SavingsAccount(BankAccount):
    def calculate_interest(self, rate):
        interest = self.balance * rate / 100
        print("Interest:", interest)


a = SavingsAccount(10000)
a.deposit(2000)
a.withdraw(1000)
a.calculate_interest(5)
print("Balance:", a.balance)
```

```python
# 7. Product → Electronics

class Product:
    def __init__(self, price):
        self.price = price

    def discount(self, percentage):
        discount_amount = self.price * percentage / 100
        return self.price - discount_amount


class Electronics(Product):
    def __init__(self, price, warranty):
        super().__init__(price)
        self.warranty = warranty

    def display(self):
        print("Price:", self.price)
        print("Warranty:", self.warranty, "years")


e = Electronics(30000, 2)
e.display()
print("Price after 10% discount:", e.discount(10))
```

```python
# 12. Account → LoanAccount

class Account:
    def __init__(self, balance):
        self.balance = balance


class LoanAccount(Account):
    def calculate_emi(self, principal, rate, months):
        monthly_rate = rate / (12 * 100)

        emi = (principal * monthly_rate * (1 + monthly_rate) ** months) / \
              ((1 + monthly_rate) ** months - 1)

        print("EMI:", round(emi, 2))


l = LoanAccount(50000)
l.calculate_emi(100000, 10, 12)
```

```python
# 29. Shape → TwoDShape → Triangle

import math


class Shape:
    def display(self):
        print("This is a shape")


class TwoDShape(Shape):
    def dimensions(self):
        print("This is a two-dimensional shape")


class Triangle(TwoDShape):
    def __init__(self, side1, side2, side3):
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3

    def area(self):
        s = (self.side1 + self.side2 + self.side3) / 2

        area = math.sqrt(
            s * (s - self.side1) *
            (s - self.side2) *
            (s - self.side3)
        )

        print("Area of Triangle:", area)


t = Triangle(3, 4, 5)
t.display()
t.dimensions()
t.area()
```

```python
# 30. GrandParent → Parent → Child → GrandChild

class GrandParent:
    def grandparent_method(self):
        print("GrandParent method")


class Parent(GrandParent):
    def parent_method(self):
        print("Parent method")


class Child(Parent):
    def child_method(self):
        print("Child method")


class GrandChild(Child):
    def grandchild_method(self):
        print("GrandChild method")


g = GrandChild()

g.grandparent_method()
g.parent_method()
g.child_method()
g.grandchild_method()

print("\nMethod Resolution Order:")
print(GrandChild.mro())
```
