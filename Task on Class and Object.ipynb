```python
# 26. Temperature — Getter/Setter with Conversion

class Temperature:
    def __init__(self, celsius):
        self.__celsius = celsius

    def get_celsius(self):
        return self.__celsius

    def set_celsius(self, celsius):
        self.__celsius = celsius

    def get_fahrenheit(self):
        return (self.__celsius * 9 / 5) + 32


temperature = Temperature(25)

print("Celsius:", temperature.get_celsius())
print("Fahrenheit:", temperature.get_fahrenheit())

temperature.set_celsius(30)

print("Celsius:", temperature.get_celsius())
print("Fahrenheit:", temperature.get_fahrenheit())
```

```python
# 28. Product — Price and Discount

class Product:
    def __init__(self, price):
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price > 0:
            self.__price = price
        else:
            print("Invalid price")

    def apply_discount(self, discount):
        self.__price = self.__price - (
            self.__price * discount / 100
        )


product = Product(1000)

print("Original price:", product.get_price())

product.apply_discount(10)

print("After discount:", product.get_price())

product.set_price(2000)

print("New price:", product.get_price())
```

```python
# Part E — Hard OOP Questions

# 1. Employee — Salary Validation + Class Attribute

class Employee:
    total_employees = 0
    last_updated_salary = None

    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

        Employee.total_employees += 1
        Employee.last_updated_salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):

        if new_salary < 5000:
            print("Salary cannot be less than 5000")
            return

        maximum_salary = self.__salary * 1.20

        if new_salary > maximum_salary:
            print("Salary cannot increase by more than 20%")
            return

        self.__salary = new_salary

        Employee.last_updated_salary = new_salary

        print("Salary updated:", new_salary)

    @classmethod
    def get_total_employees(cls):
        return cls.total_employees


emp1 = Employee("Ram", 50000)
emp2 = Employee("Sita", 60000)

print("Total employees:",
      Employee.get_total_employees())

emp1.set_salary(55000)   # Valid
emp1.set_salary(70000)   # More than 20%
emp1.set_salary(4000)    # Less than 5000

print("Ram salary:", emp1.get_salary())
print("Last updated salary:",
      Employee.last_updated_salary)
```

```python
# 2. BankAccount — Minimum Balance, Bonus & History

class BankAccount:
    total_accounts = 0
    history = []

    def __init__(self, balance):
        if balance >= 1000:
            self.__balance = balance
        else:
            self.__balance = 1000
            print("Minimum balance is 1000")

        BankAccount.total_accounts += 1
        BankAccount.history.append(self.__balance)

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):

        if balance < 1000:
            print("Balance cannot go below 1000")
            return

        if balance > 100000:
            bonus = balance * 5 / 100
            balance = balance + bonus
            print("5% bonus added")

        self.__balance = balance

        BankAccount.history.append(self.__balance)

        print("Balance updated:", self.__balance)

    @classmethod
    def get_total_accounts(cls):
        return cls.total_accounts

    @classmethod
    def show_history(cls):
        print("Balance history:", cls.history)


account1 = BankAccount(50000)
account2 = BankAccount(120000)

account1.set_balance(60000)
account1.set_balance(500)

print("Balance:", account1.get_balance())

print("Total accounts:",
      BankAccount.get_total_accounts())

BankAccount.show_history()
```

```python
# 3. Student — Marks List, Average and Status

class Student:
    total_students = 0

    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks

        if any(mark < 40 for mark in marks):
            self.status = "Failed"
        else:
            self.status = "Passed"

        Student.total_students += 1

    def get_marks(self):
        return self.__marks

    def get_average(self):
        return sum(self.__marks) / len(self.__marks)

    def set_marks(self, new_marks):

        # Check number of subjects
        if len(new_marks) != 5:
            print("Enter marks for exactly 5 subjects")
            return

        # Check range
        for mark in new_marks:
            if mark < 0 or mark > 100:
                print("Marks must be between 0 and 100")
                return

        old_average = self.get_average()
        new_average = sum(new_marks) / len(new_marks)

        if new_average < old_average:
            print("New average cannot be lower")
            return

        self.__marks = new_marks

        if any(mark < 40 for mark in new_marks):
            self.status = "Failed"
        else:
            self.status = "Passed"

        print("Marks updated successfully")


student = Student(
    "Ram",
    [80, 85, 90, 75, 88]
)

print("Marks:", student.get_marks())
print("Average:", student.get_average())
print("Status:", student.status)

student.set_marks([85, 90, 95, 80, 90])

print("Marks:", student.get_marks())
print("Average:", student.get_average())
print("Status:", student.status)

print("Total students:",
      Student.total_students)
```

```python
# 4. Product — Price, Discount & Revenue

class Product:
    total_revenue = 0

    def __init__(self, product_name, price):
        self.product_name = product_name
        self.__price = price
        self.__discount = 0

        Product.total_revenue += price

    def get_price(self):
        return self.__price

    def set_price(self, new_price):

        minimum_price = self.__price * 0.80

        if new_price < minimum_price:
            print("Price cannot be below 80% of current price")
            return

        old_price = self.__price

        self.__price = new_price

        if new_price > 5000:
            self.__discount = 15
        else:
            self.__discount = 0

        Product.total_revenue += new_price - old_price

        print("Price updated:", self.__price)
        print("Discount:", self.__discount, "%")

    def get_discount(self):
        return self.__discount


product = Product("Laptop", 6000)

print("Price:", product.get_price())

product.set_price(5500)

print("Price:", product.get_price())
print("Discount:", product.get_discount(), "%")

product.set_price(3000)

print("Total revenue:",
      Product.total_revenue)
```

```python
# 5. User — Password, Age & Senior Citizen

class User:
    senior_citizens = 0

    def __init__(self, password, age):
        self.__password = password
        self.__age = age
        self.senior_citizen = False

        if age >= 60:
            self.senior_citizen = True
            User.senior_citizens += 1

    def get_password(self):
        return self.__password

    def get_age(self):
        return self.__age

    def set_password(self, password):

        if len(password) < 8:
            print("Password must contain at least 8 characters")
            return False

        if not any(char.isdigit() for char in password):
            print("Password must contain at least one digit")
            return False

        self.__password = password
        print("Password updated")
        return True

    def set_age(self, age):

        if age < self.__age:
            print("Age cannot be decreased")
            return False

        if age >= 60 and self.__age < 60:
            self.senior_citizen = True
            User.senior_citizens += 1

        self.__age = age

        print("Age updated:", self.__age)

        return True

    def update(self, password, age):

        # First validate password
        password_valid = self.set_password(password)

        # Then validate age
        age_valid = self.set_age(age)

        if password_valid and age_valid:
            print("User updated successfully")
        else:
            print("Some updates failed")


user = User("abc12345", 55)

print("Age:", user.get_age())

user.update("newpass123", 60)

print("Age:", user.get_age())
print("Senior citizen:", user.senior_citizen)

print("Total senior citizens:",
      User.senior_citizens)
```
