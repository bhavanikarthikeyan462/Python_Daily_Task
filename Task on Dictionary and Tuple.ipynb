```python
# 2. print all keys using keys()

d = {"a": 1, "b": 2, "c": 3}

print(d.keys())
```

```python
# 4. get value safely using get()

d = {
    "name": "Sumit",
    "age": 25
}

name = d.get("name")

print("Name:", name)
```

```python
# 6. count 20 in a tuple

tup = (10, 20, 30, 40, 50)

count = tup.count(20)

print("20 appears:", count, "time")
```

```python
# 7. find the index of 30

tup = (10, 20, 30, 40)

position = tup.index(30)

print("Index of 30:", position)
```

```python
# 8. create an empty dictionary and add 3 pairs

student = {}

student["name"] = "Rahul"
student["age"] = 22
student["city"] = "Hyderabad"

print(student)
```

```python
# 9. add items using update()

d = {"a": 1, "b": 2}

d.update({"c": 3, "d": 4})

print(d)
```

```python
# 10. remove age using pop()

student = {
    "name": "Rahul",
    "age": 22,
    "city": "Delhi"
}

removed = student.pop("age")

print("Removed:", removed)
print("Dictionary:", student)
```

```python
# 11. use popitem()

d = {
    "x": 10,
    "y": 20,
    "z": 30
}

removed = d.popitem()

print("Removed item:", removed)
print("Dictionary:", d)
```

```python
# 12. copy a dictionary using copy()

original = {
    "a": 1,
    "b": 2
}

copy_dict = original.copy()

print("Original:", original)
print("Copy:", copy_dict)
```

```python
# 13. create dictionary using fromkeys()

keys = ["red", "green", "blue"]

d = dict.fromkeys(keys, 0)

print(d)
```

```python
# 14. try to change a tuple element

tup = ("Python", 25, 10.5, True)

print("Original tuple:", tup)

tup[1] = 30
```

```python
# 15. use setdefault()

d = {
    "name": "Sumit"
}

d.setdefault("age", 25)

print(d)
```

```python
# 16. Add 3 items using update() and remove last item using popitem()

data = {}

data.update({"a": 10})
data.update({"b": 20})
data.update({"c": 30})

print("Before popitem:", data)

removed = data.popitem()

print("Removed item:", removed)
print("After popitem:", data)
```

```python
# 17. Find index and count a number in a tuple

numbers = (10, 20, 30, 20, 40, 20)

number = 20

print("Index of", number, ":", numbers.index(number))
print("Count of", number, ":", numbers.count(number))
```

```python
# 18. Take 3 student names and marks as input

students = {}

for i in range(3):
    name = input("Enter student name: ")
    mark = int(input("Enter marks: "))
    
    students[name] = mark

print("Student Dictionary:", students)
```

```python
# 19. Print keys where value is 10

data = {
    "a": 10,
    "b": 20,
    "c": 10
}

for key, value in data.items():
    if value == 10:
        print(key)
```

```python
# 20. Copy dictionary, clear original, and show both

data = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40,
    "e": 50
}

copy_data = data.copy()

data.clear()

print("Original:", data)
print("Copy:", copy_data)
```

```python
# 21. Update dictionary, remove last item, print all items

student = {
    "name": "Rahul",
    "age": 22,
    "city": "Hyderabad",
    "course": "Python"
}

new_data = {
    "marks": 85,
    "grade": "A"
}

student.update(new_data)

print("Before popitem:", student)

removed = student.popitem()

print("Removed item:", removed)

print("All items:")

for key, value in student.items():
    print(key, ":", value)
```

```python
# 22. Find 25, count it, convert tuple to list and sort descending

numbers = (10, 25, 30, 20, 25, 40, 25, 50)

if 25 in numbers:
    print("Index of 25:", numbers.index(25))
    print("Count of 25:", numbers.count(25))
else:
    print("25 does not exist")

numbers_list = list(numbers)

numbers_list.sort(reverse=True)

print("Descending order:", numbers_list)
```

```python
# 23. Find student with highest marks

students = {}

for i in range(4):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    
    students[name] = marks

highest_student = max(students, key=students.get)

print("Students:", students)
print("Highest Marks Student:", highest_student)
print("Marks:", students[highest_student])
```

```python
# 24. Copy, clear original, and update copy

info = {
    "name": "Aman",
    "age": 21,
    "city": "Mumbai"
}

copy_info = info.copy()

info.clear()

copy_info["country"] = "India"

print("Original:", info)
print("Copy:", copy_info)
```

```python
# 25. Count tuple elements using dictionary

data = (10, 20, 30, 20, 40, 20, 50)

count_dict = {}

for number in data:
    count_dict[number] = data.count(number)

print(count_dict)
```
