```python
# 1. Division with ZeroDivisionError

try:
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    result = a / b
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
# 6. Multiple exceptions – ValueError and ZeroDivisionError

try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    print("Result:", a / b)

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
# 7. else block with try-except

try:
    num = int(input("Enter a number: "))
    result = 100 / num

except ValueError:
    print("Invalid input.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)
    print("No exception occurred.")
```

```python
# 8. finally block

try:
    num = int(input("Enter a number: "))
    print("Number:", num)

except ValueError:
    print("Invalid input.")

finally:
    print("This block always executes.")
```

# Raising Exceptions & Custom Exceptions (19–25)

```python
# 19. Raise ValueError if age is less than 18

def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or above.")

    print("Eligible.")


try:
    age = int(input("Enter your age: "))
    check_age(age)

except ValueError as e:
    print("Error:", e)
```

```python
# 20. Custom NegativeNumberError

class NegativeNumberError(Exception):
    pass


try:
    num = int(input("Enter a number: "))

    if num < 0:
        raise NegativeNumberError("Negative numbers are not allowed.")

    print("Number:", num)

except NegativeNumberError as e:
    print("Error:", e)
```

```python
# 21. Custom InsufficientBalanceError

class InsufficientBalanceError(Exception):
    pass


balance = 5000

try:
    amount = float(input("Enter withdrawal amount: "))

    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance.")

    balance -= amount
    print("Withdrawal successful.")
    print("Remaining balance:", balance)

except InsufficientBalanceError as e:
    print("Error:", e)
```

```python
# 22. Manually raise TypeError

try:
    value = input("Enter a string: ")

    if not isinstance(value, str):
        raise TypeError("Input must be a string.")

    print("You entered:", value)

except TypeError as e:
    print("Error:", e)
```

```python
# 23. Custom exception for short password

class ShortPasswordError(Exception):
    pass


def check_password(password):
    if len(password) < 8:
        raise ShortPasswordError(
            "Password must contain at least 8 characters."
        )

    print("Password accepted.")


try:
    password = input("Enter password: ")
    check_password(password)

except ShortPasswordError as e:
    print("Error:", e)
```

```python
# 28. Banking system with custom exception

class InsufficientBalanceError(Exception):
    pass


class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount.")
        else:
            self.balance += amount
            print("Deposited:", amount)

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientBalanceError(
                "Insufficient balance."
            )

        if amount <= 0:
            print("Invalid withdrawal amount.")
        else:
            self.balance -= amount
            print("Withdrawn:", amount)

    def show_balance(self):
        print("Balance:", self.balance)


account = BankAccount(5000)

try:
    account.deposit(2000)
    account.withdraw(8000)

except InsufficientBalanceError as e:
    print("Error:", e)

finally:
    account.show_balance()
```

```python
# 29. Convert list of strings to integers

try:
    values = input("Enter numbers separated by spaces: ")

    numbers = values.split()

    numbers = [int(x) for x in numbers]

    print("Integer list:", numbers)

except ValueError:
    print("Invalid input. Please enter numbers only.")
```

```python
# 30. Complete program – name, age and marks

try:
    name = input("Enter your name: ")

    age = int(input("Enter your age: "))

    marks = float(input("Enter your marks: "))

    if name == "":
        raise ValueError("Name cannot be empty.")

    if age < 0:
        raise ValueError("Age cannot be negative.")

    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    print("\nStudent Details")
    print("Name:", name)
    print("Age:", age)
    print("Marks:", marks)

except ValueError as e:
    print("Value Error:", e)

except TypeError as e:
    print("Type Error:", e)

except Exception as e:
    print("Some other error occurred:", e)

finally:
    print("Program Ended")
```
