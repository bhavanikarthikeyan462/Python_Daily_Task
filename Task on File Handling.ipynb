# File Handling

```python
# create a txt file and write your name and age

file = open("Student.txt", "w")
file.write("Name : Bhavani\n")
file.write("Age : 24")
file.close()
print("Data written Successfully.")
```

```python
# read the content of a text file

file = open("student.txt", "r")
print(file.read())
file.close()
```

```python
# Append some text to an existing file

file = open("student.txt", "a")
file.write("\nCourse: Python")
file.close()
print("Text Appended Successfully")
```

```python
# count the number of lines

file = open("Student.txt", "r")
lines = file.readlines()
print("Number of lines:", len(lines))
file.close()
```

```python
# count the number of words

file = open("Student.txt", "r")
text = file.read()
words = text.split()
print("Number of Words:", len(words))
file.close()
```

```python
# count the number of characters

file = open("Student.txt", "r")
text = file.read()
print("Number of Characters:", len(text))
file.close()
```

```python
# Read a file line by line using for loop

file = open("Student.txt", "r")
for line in file:
  print(line.strip())
file.close()
```

```python
# copy content from one file to another file

source = open("student.txt", "r")
data = source.read()
source.close()

target = open("copy.txt", "w")
target.write(data)
target.close()

print("File copied successfully.")
```

```python
# check if a file exists

import os
if os.path.exists("student.txt"):
    file = open("student.txt", "r")
    print(file.read())
    file.close()
else:
    print("File does not exist.")
```

```python
# delete a file (use os module)

import os
if os.path.exists("copy.text"):
    os.remove("copy.txt")
    print("File Deleted")
else:
    print("File not found")
```

```python
# Take student name and marks from user and store them in a file

name = input("Enter student name:")
marks = input("Enter marks:")

file = open("marks.txt", "a")
file.write(name + " " + marks + "\n")
file.close()

print("Data Stored Successfully")
```

```python
# read a file and print only the lines that contain the word “Python”

file = open("student.txt", "r")
for line in file:
    if "Python" in line:
        print(line.strip())
file.close()
```

```python
# reverse the content of a file and save it in another file

file = open("Student.txt", "r")
text = file.read()
file.close()

reverse = text[::-1]

newfile = open("reverse.txt", "w")
newfile.write(reverse)
newfile.close()

print("Reversed content saved.")
```

```python
# Read a file and display the longest word in it

file = open("Student.txt", "r")
text = file.read()

words = text.split()

longest = max(words, key=len)
print("Longest word:", longest)
file.close()
```

```python
# merge two text files into a third file

file1 = open("student.txt", "r")
file2 = open("copy.txt", "r")

data1 = file1.read()
data2 = file2.read()

file3 = open("merged.txt", "w")
file3.write(data1)
file3.write("\n")
file3.write(data2)

file1.close()
file2.close()
file3.close()

print("Files merged successfully.")
```

```python
# Store a list of numbers in a file, then read and calculate their sum

file = open("numbers.txt", "w")
file.write("10\n20\n30\n40\n50")
file.close()

file = open("numbers.txt", "r")
numbers = file.readlines()

total = 0
for num in numbers:
    total ++ int(num.strip())

print("sum =", total)

file.close()
```
