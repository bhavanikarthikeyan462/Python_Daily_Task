```python
# 5 Hard Questions on Python Functions
#1. Hard — *args + **kwargs
#Question:

#Write a function process_data(*args, **kwargs) that:

#Takes any number of positional arguments (numbers).
#Takes keyword arguments where keys are operation names ("add", "multiply", "power").
#Performs the requested operations on all numbers.
#Returns a dictionary with results.

def process_data(*args, **kwargs):
    result = {}

    if not args:
        return result

    if kwargs.get("add"):
        result["add"] = sum(args)

    if kwargs.get("multiply"):
        product = 1

        for num in args:
            product *= num

        result["multiply"] = product

    if kwargs.get("power"):
        result["power"] = [num ** 2 for num in args]

    return result


print(process_data(2, 3, 4, add=True, multiply=True))
```

```python
2.# Create a function make_multiplier(n) that returns another function which multiplies any number by n.

def make_multiplier(n):

    def multiplier(x):
        return x * n

    return multiplier


double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))
print(triple(5))
```

```python
3.# Write a function smart_divide(func) as a decorator that:

#Takes a function as an argument.
#Checks if the second argument is zero before division.
#If zero, prints "Cannot divide by zero!" and returns None.
#Otherwise, calls the original function.

def smart_divide(func):

    def wrapper(a, b):
        if b == 0:
            print("Cannot divide by zero!")
            return None

        return func(a, b)

    return wrapper


@smart_divide
def divide(a, b):
    return a / b


print(divide(10, 2))
print(divide(10, 0))
```

```python
4. # Takes any number of numbers.
# Supports add, subtract, multiply, and divide.
# If round_result=True is present in kwargs, rounds the final answer to 2 decimal places.
# Handles edge cases.

def flexible_calc(*args, operation="add", **kwargs):

    round_result = kwargs.get("round_result", False)

    if not args:
        return None

    if not all(isinstance(x, (int, float)) for x in args):
        return None

    if operation == "add":
        result = sum(args)

    elif operation == "subtract":
        result = args[0]
        for x in args[1:]:
            result -= x

    elif operation == "multiply":
        result = 1
        for x in args:
            result *= x

    elif operation == "divide":
        if any(x == 0 for x in args[1:]):
            print("Cannot divide by zero!")
            return None

        result = args[0]
        for x in args[1:]:
            result /= x

    else:
        print("Invalid operation!")
        return None

    if round_result:
        result = round(result, 2)

    return result

print(flexible_calc(10, 20, 30))
```

```python
# 5. Write a recursive function flatten_list(nested_list) that takes a nested list and returns a single flat list.

def flatten_list(nested_list):
    result = []

    for item in nested_list:
        if isinstance(item, list):
            result.extend(flatten_list(item))
        else:
            result.append(item)

    return result


data = [1, [2, 3], [4, [5, 6]]]

print(flatten_list(data))
```
