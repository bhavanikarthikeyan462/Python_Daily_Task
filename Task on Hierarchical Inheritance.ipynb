```python
# Hierarchical Inheritance with Constructors & Methods


# 13. Person Constructor

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks


class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject


s = Student("Rahul", 20, 85)
t = Teacher("Anita", 35, "Python")

print(s.name, s.age, s.marks)
print(t.name, t.age, t.subject)
```

```python
# 14. Animal Sound Override

class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")


d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
# 15. Vehicle Constructor

class Vehicle:
    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors


class Bike(Vehicle):
    def __init__(self, brand, engine_cc):
        super().__init__(brand)
        self.engine_cc = engine_cc


c = Car("Toyota", 4)
b = Bike("Honda", 150)

print("Car:", c.brand, c.doors)
print("Bike:", b.brand, b.engine_cc)
```

```python
# 16. Shape Area

class Shape:
    def area(self):
        print("Area")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print("Circle Area:", 3.14 * self.radius * self.radius)


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        print("Square Area:", self.side * self.side)


c = Circle(5)
s = Square(4)

c.area()
s.area()
```

```python
# 17. Employee Details

class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def get_details(self):
        print("Name:", self.name)
        print("Employee ID:", self.emp_id)


class Manager(Employee):
    def __init__(self, name, emp_id, team):
        super().__init__(name, emp_id)
        self.team = team

    def get_details(self):
        super().get_details()
        print("Team:", self.team)


class Developer(Employee):
    def __init__(self, name, emp_id, language):
        super().__init__(name, emp_id)
        self.language = language

    def get_details(self):
        super().get_details()
        print("Language:", self.language)


m = Manager("Rahul", 101, 10)
d = Developer("Amit", 102, "Python")

m.get_details()
d.get_details()
```

```python
# 23. School Management

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def display(self):
        print("Student:", self.name, self.age, self.marks)


class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def display(self):
        print("Teacher:", self.name, self.age, self.subject)


class Staff(Person):
    def __init__(self, name, age, department):
        super().__init__(name, age)
        self.department = department

    def display(self):
        print("Staff:", self.name, self.age, self.department)


s = Student("Rahul", 20, 85)
t = Teacher("Anita", 35, "Python")
st = Staff("Ramesh", 40, "Office")

s.display()
t.display()
st.display()
```

```python
# 24. Bank System

class Account:
    def __init__(self, balance):
        self.balance = balance


class Savings(Account):
    def interest(self):
        print("Savings Interest:", self.balance * 0.05)


class Current(Account):
    def overdraft(self):
        print("Overdraft Limit: 20000")


class FixedDeposit(Account):
    def maturity(self):
        print("Maturity Amount:", self.balance * 1.10)


s = Savings(10000)
c = Current(20000)
f = FixedDeposit(50000)

s.interest()
c.overdraft()
f.maturity()
```

```python
# 27. Employee Salary

class Employee:
    def __init__(self, name):
        self.name = name

    def calculate_salary(self):
        print("Salary calculation")


class FullTime(Employee):
    def __init__(self, name, monthly_salary):
        super().__init__(name)
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        print(self.name, "Monthly Salary:", self.monthly_salary)


class PartTime(Employee):
    def __init__(self, name, hours, rate):
        super().__init__(name)
        self.hours = hours
        self.rate = rate

    def calculate_salary(self):
        print(self.name, "Hourly Salary:", self.hours * self.rate)


f = FullTime("Rahul", 50000)
p = PartTime("Amit", 20, 500)

f.calculate_salary()
p.calculate_salary()
```

```python
# 30. Your Own Hierarchical Design

# Example: Vehicle → Car, Bike, Bus

class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def start(self):
        print(self.brand, self.model, "started")


class Car(Vehicle):
    def open_door(self):
        print("Car door opened")


class Bike(Vehicle):
    def wear_helmet(self):
        print("Wear helmet before riding")


class Bus(Vehicle):
    def open_door(self):
        print("Bus door opened")


car = Car("Toyota", "Innova")
bike = Bike("Honda", "Shine")
bus = Bus("Volvo", "Bus 500")

# Parent attributes and method
print(car.brand, car.model)
car.start()

# Child-specific method
car.open_door()

print()

print(bike.brand, bike.model)
bike.start()
bike.wear_helmet()

print()

print(bus.brand, bus.model)
bus.start()
bus.open_door()
```
