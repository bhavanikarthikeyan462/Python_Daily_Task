# Task on Hybrid Inheritance

```python
# 1. A → B and C → D

class A:
    def method_a(self):
        print("Method of A")


class B(A):
    def method_b(self):
        print("Method of B")


class C:
    def method_c(self):
        print("Method of C")


class D(B, C):
    def method_d(self):
        print("Method of D")


obj = D()

obj.method_a()
obj.method_b()
obj.method_c()
obj.method_d()
```

```python
# 2. Person, Employee, Student, TeachingAssistant

class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary


class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks


class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        Person.__init__(self, name)
        self.salary = salary
        self.marks = marks
        self.department = department


ta = TeachingAssistant("Rahul", 30000, 85, "Computer Science")

print("Name:", ta.name)
print("Salary:", ta.salary)
print("Marks:", ta.marks)
print("Department:", ta.department)
```

```python
# 3. Vehicle → Car, Electric → ElectricCar

class Vehicle:
    def start(self):
        print("Vehicle starts")


class Car(Vehicle):
    def drive(self):
        print("Car is driving")


class Electric:
    def battery(self):
        print("Battery is charging")


class ElectricCar(Car, Electric):
    pass


obj = ElectricCar()

obj.start()
obj.drive()
obj.battery()
```

```python
# 4. Demonstrate Hybrid Inheritance and MRO

class A:
    def show(self):
        print("A")


class B(A):
    def show(self):
        print("B")


class C:
    def show(self):
        print("C")


class D(B, C):
    def show(self):
        print("D")


obj = D()

obj.show()

print(D.mro())
```

```python
# 5. Diamond Problem using super()

class A:
    def __init__(self):
        print("A constructor")
        super().__init__()


class B(A):
    def __init__(self):
        print("B constructor")
        super().__init__()


class C(A):
    def __init__(self):
        print("C constructor")
        super().__init__()


class D(B, C):
    def __init__(self):
        print("D constructor")
        super().__init__()


obj = D()
```

```python
# 6. Banking System using Hybrid Inheritance

class Account:
    def __init__(self, account_no):
        self.account_no = account_no


class Savings(Account):
    def __init__(self, account_no, balance):
        super().__init__(account_no)
        self.balance = balance


class Loan(Account):
    def __init__(self, account_no, loan_amount):
        super().__init__(account_no)
        self.loan_amount = loan_amount


class SavingsLoanAccount(Savings, Loan):
    def __init__(self, account_no, balance, loan_amount):
        Account.__init__(self, account_no)
        self.balance = balance
        self.loan_amount = loan_amount


obj = SavingsLoanAccount("ACC101", 50000, 20000)

print("Account Number:", obj.account_no)
print("Balance:", obj.balance)
print("Loan Amount:", obj.loan_amount)
```

```python
# 7. Shape, Color, Circle, ColoredCircle

class Shape:
    def area(self):
        print("This is the area method")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


class Color:
    def color(self):
        print("Color is Red")


class ColoredCircle(Circle, Color):
    pass


obj = ColoredCircle(5)

print("Area:", obj.area())
obj.color()
```

```python
# 8. Constructor Chaining

class A:
    def __init__(self):
        print("A constructor")
        super().__init__()


class B(A):
    def __init__(self):
        print("B constructor")
        super().__init__()


class C(A):
    def __init__(self):
        print("C constructor")
        super().__init__()


class D(B, C):
    def __init__(self):
        print("D constructor")
        super().__init__()


obj = D()

print("MRO:", D.mro())
```

```python
# 9. Method Overriding in Hybrid Inheritance

class A:
    def show(self):
        print("Show method of A")


class B(A):
    def show(self):
        print("Show method of B")


class C(A):
    def show(self):
        print("Show method of C")


class D(B, C):
    pass


obj = D()

obj.show()

print(D.mro())
```

```python
# Duck Typing — Questions 11–20
# 11. What is Duck Typing?
# Duck typing means that Python cares about what an object can do, rather than what class the object belongs to.

class Dog:
    def speak(self):
        print("Dog says Woof")


class Robot:
    def speak(self):
        print("Robot says Hello")


def make_speak(obj):
    obj.speak()


dog = Dog()
robot = Robot()

make_speak(dog)
make_speak(robot)
```

```python
# 15. Circle and Rectangle with area()

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


def calculate(obj):
    print("Area:", obj.area())


circle = Circle(5)
rectangle = Rectangle(10, 4)

calculate(circle)
calculate(rectangle)
```

```python
# 18. FileLogger and ConsoleLogger

class FileLogger:
    def log(self, message):
        print("File Log:", message)


class ConsoleLogger:
    def log(self, message):
        print("Console Log:", message)


def write_log(logger, message):
    logger.log(message)


file_logger = FileLogger()
console_logger = ConsoleLogger()

write_log(file_logger, "Error occurred")
write_log(console_logger, "Program started")
```

```python
# 19. Hybrid Inheritance + Duck Typing

class A:
    def show(self):
        print("Show method of A")


class B(A):
    pass


class C(A):
    pass


class D(B, C):
    def show(self):
        print("Show method of D")


class Other:
    def show(self):
        print("Show method of Other")


def display(obj):
    obj.show()


d = D()
other = Other()

display(d)
display(other)
```

```python
# 20. Employee Hybrid Inheritance + Duck Typing

class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    pass


class Manager(Employee):
    pass


class HR(Person):
    pass


class TeamLead(Manager, HR):
    def work(self):
        print(self.name, "is working as Team Lead")


class Freelancer:
    def __init__(self, name):
        self.name = name

    def work(self):
        print(self.name, "is working as Freelancer")


def do_work(obj):
    obj.work()


team_lead = TeamLead("Rahul")
freelancer = Freelancer("Amit")

do_work(team_lead)
do_work(freelancer)
```
