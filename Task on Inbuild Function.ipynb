```python
# 3. Advanced Enumeration + Zip + Condition

names = ["Sumit", "Rahul", "Priya", "Aman", "Neha"]
scores = [85, 92, 78, 95, 88]
```

```python
result = [
    (i, name, "Excellent" if score >= 90 else "Good")
    for i, (name, score) in enumerate(zip(names, scores))
]
result
```

```python
# 4. Very Hard — Nested + Multiple Built-ins

# Question: Write a list comprehension to generate a multiplication table (5×5) where:

result = [
    [
        str(i * j) + (" (ok)" if i * j % 2 == 0 else " (odd)")
        for j in range(1, 6)
        if i * j % 7 != 0
    ]
    for i in range(1, 6)
]
result
```
