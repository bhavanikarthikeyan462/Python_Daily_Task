```python
# stopiteration

numbers = [10,20,30]

it = iter(numbers)

try:
    print(next(it))
    print(next(it))
    print(next(it))
    print(next(it))
except StopIteration:
    print("No more elements in iterator")
```

```python
# Create a custom iterator that prints numbers from 1 to 10

class Numbers:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

obj = Numbers()

for i in obj:
    print(i)
```

```python
# Create a custom iterator that prints only even numbers from 2 to 20

class EvenNumbers:

    def __init__(self):
        self.num = 2

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 20:
            value = self.num
            self.num += 2
            return value
        else:
            raise StopIteration

obj = EvenNumbers()

for i in obj:
    print(i)
```

```python
# Create a custom iterator that prints numbers in reverse order from 10 to 1

class Reverse:

    def __init__(self):
        self.num = 10

    def __iter__(self):
        return self

    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value
        else:
            raise StopIteration

obj = Reverse()

for i in obj:
    print(i)
```

```python
# Create a custom iterator that prints the characters of the word "GENERATOR" one by one


class Word:
    def __init__(self):
        self.word = "GENERATOR"
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.word):
            ch = self.word[self.index]
            self.index += 1
            return ch
        else:
            raise StopIteration

obj = Word()

for i in obj:
    print(i)
```

```python
# Create a custom iterator that prints the multiplication table of 5

class Table:

    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 0:
            value = 5 * self.num
            self.num += 1
            return value
        else:
            raise StopIteration

obj = Table()

for i in obj:
    print(i)
```
