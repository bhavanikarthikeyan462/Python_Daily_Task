# Task on Lambda Function

```python
# 1. Write a lambda function to return the square of a number.

square = lambda x: x * x

print(square(5))
```

```python
# 2. Write a lambda function to return the cube of a number.

cube = lambda x: x ** 3

print(cube(4))
```

```python
# 3. Write a lambda function to add two numbers.

add = lambda a, b: a + b

print(add(10, 20))
```

```python
# 4. Write a lambda function to find the maximum of two numbers.

maximum = lambda a, b: max(a, b)

print(maximum(15, 8))
```

```python
# 5. Write a lambda function that returns "Even" if a number is even; otherwise "Odd".

check = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check(9))
```

```python
# 6. Use map() with lambda to square every element of a list.

numbers = [1, 2, 3, 4]

result = list(map(lambda x: x ** 2, numbers))

print(result)
```

```python
# 7. Use filter() with lambda to keep only even numbers.

numbers = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
```

```python
# 8. Use filter() with lambda to keep numbers greater than 50.

numbers = [20, 45, 55, 70]

result = list(filter(lambda x: x > 50, numbers))

print(result)
```

```python
# 9. Sort a list of numbers in ascending order using lambda.

numbers = [9, 5, 1, 8, 2]

result = sorted(numbers, key=lambda x: x)

print(result)
```

```python
# 10. Sort a list of numbers in descending order using lambda.

numbers = [9, 5, 1, 8, 2]

result = sorted(numbers, key=lambda x: x, reverse=True)

print(result)
```

```python
# 11. Use reduce() with lambda to find the sum of all numbers.

from functools import reduce

numbers = [10, 20, 30, 40]

result = reduce(lambda a, b: a + b, numbers)

print(result)
```

```python
# 12. Use reduce() with lambda to find the product of all numbers.

from functools import reduce

numbers = [2, 3, 4, 5]

result = reduce(lambda a, b: a * b, numbers)

print(result)
```

```python
# 30. Interview Challenge — Student Analysis

students = [
    {"name": "Ram", "marks": [70, 80, 90]},
    {"name": "Amit", "marks": [90, 92, 94]},
    {"name": "Neha", "marks": [60, 65, 70]},
    {"name": "Raj", "marks": [88, 84, 90]}
]
```

```python
# 1. Calculate each student's average marks.

result = list(map(
    lambda s: {
        "name": s["name"],
        "average": sum(s["marks"]) / len(s["marks"])
    },
    students
))

print(result)
```

```python
# 2. Sort students by average marks, highest first.

result = sorted(
    students,
    key=lambda s: sum(s["marks"]) / len(s["marks"]),
    reverse=True
)

print(result)
```

```python
# 3. Find the topper.

topper = max(
    students,
    key=lambda s: sum(s["marks"]) / len(s["marks"])
)

print(topper)
```

```python
# 4. Filter students whose average is greater than 80.

result = list(filter(
    lambda s: sum(s["marks"]) / len(s["marks"]) > 80,
    students
))

print(result)
```

```python
# 5. Print only the names of filtered students.

result = list(map(
    lambda s: s["name"],
    filter(
        lambda s: sum(s["marks"]) / len(s["marks"]) > 80,
        students
    )
))

print(result)
```

```python
# Interview Question 1 — Student Ranking Challenge

students = [
    {
        "name": "Ram",
        "marks": {"Math": 80, "Python": 95, "SQL": 88}
    },
    {
        "name": "Amit",
        "marks": {"Math": 92, "Python": 90, "SQL": 91}
    },
    {
        "name": "Neha",
        "marks": {"Math": 85, "Python": 99, "SQL": 94}
    },
    {
        "name": "Raj",
        "marks": {"Math": 78, "Python": 70, "SQL": 80}
    }
]
```

```python
# 1. Calculate every student's average.

averages = list(map(
    lambda s: {
        "name": s["name"],
        "average": sum(s["marks"].values()) / len(s["marks"])
    },
    students
))

print(averages)
```

```python
# 2. Sort by average descending.

result = sorted(
    averages,
    key=lambda x: x["average"],
    reverse=True
)

print(result)
```

```python
# 3. Find topper.

topper = max(
    students,
    key=lambda s: sum(s["marks"].values())
                  / len(s["marks"])
)

print(topper["name"])
```

```python
# 4. Find lowest scorer.

lowest = min(
    students,
    key=lambda s: sum(s["marks"].values())
                  / len(s["marks"])
)

print(lowest["name"])
```

```python
# 5. Print only students whose average > 90

result = list(filter(
    lambda s: sum(s["marks"].values()) / len(s["marks"]) > 90,
    students
))

print(result)
```

```python
# 6. Print only names.

result = list(map(
    lambda s: s["name"],
    filter(
        lambda s: sum(s["marks"].values()) / len(s["marks"]) > 90,
        students
    )
))

print(result)
```

```python
# 7. Find total marks of all students.

from functools import reduce

total = reduce(
    lambda a, b: a + b,
    map(
        lambda s: sum(s["marks"].values()),
        students
    )
)

print(total)
```

```python
# Interview Question 2 — Employee Salary Analytics

employees = [
    {"id": 101, "name": "Ram", "dept": "IT", "salary": 55000},
    {"id": 102, "name": "Amit", "dept": "HR", "salary": 40000},
    {"id": 103, "name": "Neha", "dept": "IT", "salary": 70000},
    {"id": 104, "name": "Raj", "dept": "Sales", "salary": 45000},
    {"id": 105, "name": "Anu", "dept": "IT", "salary": 85000},
    {"id": 106, "name": "Kiran", "dept": "Sales", "salary": 62000}
]
```

```python
# 1. Find highest salary employee.

highest = max(employees, key=lambda x: x["salary"])

print(highest)
```

```python
# 2. Find lowest salary employee.

lowest = min(employees, key=lambda x: x["salary"])

print(lowest)
```

```python
# 3. Find average salary.

average = sum(map(lambda x: x["salary"], employees)) / len(employees)

print(average)
```

```python
# 4. Find employees earning above average.

avg = sum(map(lambda x: x["salary"], employees)) / len(employees)

result = list(filter(
    lambda x: x["salary"] > avg,
    employees
))

print(result)
```

```python
# 5. Sort by department then salary descending.

result = sorted(
    employees,
    key=lambda x: (x["dept"], -x["salary"])
)

print(result)
```

```python
# 6. Find total IT salary.

result = sum(
    map(
        lambda x: x["salary"],
        filter(lambda x: x["dept"] == "IT", employees)
    )
)

print(result)
```

```python
# 7. Print only employee names.

result = list(map(lambda x: x["name"], employees))

print(result)
```

```python
# 8. Find second highest salary.

result = sorted(
    employees,
    key=lambda x: x["salary"],
    reverse=True
)[1]

print(result)
```

```python
# Interview Question 4 — Customer Order Processing

orders = [
    {"id": 1, "customer": "Ram", "amount": 4500, "status": "Delivered"},
    {"id": 2, "customer": "Amit", "amount": 7000, "status": "Pending"},
    {"id": 3, "customer": "Raj", "amount": 2000, "status": "Delivered"},
    {"id": 4, "customer": "Neha", "amount": 9000, "status": "Delivered"},
    {"id": 5, "customer": "Anu", "amount": 3500, "status": "Cancelled"},
    {"id": 6, "customer": "Ravi", "amount": 10000, "status": "Delivered"}
]
```

```python
# 1. Total delivered amount.

result = sum(
    map(
        lambda x: x["amount"],
        filter(lambda x: x["status"] == "Delivered", orders)
    )
)

print(result)
```

```python
# 2. Average delivered amount.

delivered = list(filter(
    lambda x: x["status"] == "Delivered",
    orders
))

result = sum(map(lambda x: x["amount"], delivered)) / len(delivered)

print(result)
```

```python
# 3. Largest delivered order.

result = max(
    filter(lambda x: x["status"] == "Delivered", orders),
    key=lambda x: x["amount"]
)

print(result)
```

```python
# 4. Pending orders.

result = list(filter(
    lambda x: x["status"] == "Pending",
    orders
))

print(result)
```

```python
# 5. Sort by amount descending.

result = sorted(
    orders,
    key=lambda x: x["amount"],
    reverse=True
)

print(result)
```

```python
# 6. Print only delivered customers.

result = list(map(
    lambda x: x["customer"],
    filter(lambda x: x["status"] == "Delivered", orders)
))

print(result)
```

```python
# 7. Count delivered orders.

result = len(list(filter(
    lambda x: x["status"] == "Delivered",
    orders
)))

print(result)
```

```python
# 8. Find customer spending maximum.

result = max(
    orders,
    key=lambda x: x["amount"]
)

print(result["customer"])
```

```python
# Ultimate Interview Question — Student Analysis

students = [
    {
        "name": "Ram",
        "age": 22,
        "marks": [80, 70, 90],
        "skills": ["Python", "SQL"]
    },
    {
        "name": "Amit",
        "age": 20,
        "marks": [95, 92, 88],
        "skills": ["Python", "Power BI", "SQL"]
    },
    {
        "name": "Neha",
        "age": 24,
        "marks": [85, 90, 80],
        "skills": ["Excel", "Python"]
    },
    {
        "name": "Raj",
        "age": 21,
        "marks": [65, 70, 72],
        "skills": ["SQL"]
    }
]
```

```python
# 1. Calculate average marks.

result = list(map(
    lambda s: {
        "name": s["name"],
        "average": sum(s["marks"]) / len(s["marks"])
    },
    students
))

print(result)
```

```python
# 2. Find topper.

topper = max(
    students,
    key=lambda s: sum(s["marks"]) / len(s["marks"])
)

print(topper["name"])
```

```python
# 3. Find lowest scorer.

lowest = min(
    students,
    key=lambda s: sum(s["marks"]) / len(s["marks"])
)

print(lowest["name"])
```

```python
# 4. Sort by average descending.

result = sorted(
    students,
    key=lambda s: sum(s["marks"]) / len(s["marks"]),
    reverse=True
)

print(list(map(lambda s: s["name"], result)))
```

```python
# 5. Print names having Python skill.

result = list(map(
    lambda s: s["name"],
    filter(lambda s: "Python" in s["skills"], students)
))

print(result)
```

```python
# 6. Print names having both Python and SQL.

result = list(map(
    lambda s: s["name"],
    filter(
        lambda s: "Python" in s["skills"] and "SQL" in s["skills"],
        students
    )
))

print(result)
```

```python
# 7. Print students older than 21.

result = list(filter(
    lambda s: s["age"] > 21,
    students
))

print(result)
```

```python
# 8. Calculate overall average of all students.

overall = sum(
    map(
        lambda s: sum(s["marks"]) / len(s["marks"]),
        students
    )
) / len(students)

print(overall)
```

```python
# 9. Find student having maximum number of skills.

result = max(
    students,
    key=lambda s: len(s["skills"])
)

print(result["name"])
```

```python
# 10. Print names in alphabetical order.

result = sorted(
    map(lambda s: s["name"], students)
)

print(result)
```

```python
# 11. Find students whose average is above overall average.

overall = sum(
    map(
        lambda s: sum(s["marks"]) / len(s["marks"]),
        students
    )
) / len(students)

result = list(map(
    lambda s: s["name"],
    filter(
        lambda s: sum(s["marks"]) / len(s["marks"]) > overall,
        students
    )
))

print(result)
```

```python
# 12. Count students having Python.

result = len(list(filter(
    lambda s: "Python" in s["skills"],
    students
)))

print(result)
```

```python
# 13. Count students having SQL.

result = len(list(filter(
    lambda s: "SQL" in s["skills"],
    students
)))

print(result)
```

```python
# 14. Print only names.

result = list(map(
    lambda s: s["name"],
    students
))

print(result)
```

```python
# 15. Return the required dictionary.

overall = sum(
    map(
        lambda s: sum(s["marks"]) / len(s["marks"]),
        students
    )
) / len(students)

topper = max(
    students,
    key=lambda s: sum(s["marks"]) / len(s["marks"])
)

python_students = list(map(
    lambda s: s["name"],
    filter(lambda s: "Python" in s["skills"], students)
))

sql_students = list(map(
    lambda s: s["name"],
    filter(lambda s: "SQL" in s["skills"], students)
))

above_average = list(map(
    lambda s: s["name"],
    filter(
        lambda s: sum(s["marks"]) / len(s["marks"]) > overall,
        students
    )
))

result = {
    "Topper": topper["name"],
    "Average": overall,
    "Python Students": python_students,
    "SQL Students": sql_students,
    "Above Average": above_average,
    "Total Students": len(students)
}

print(result)
```
