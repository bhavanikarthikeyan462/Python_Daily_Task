```python
# 1. Take age and marks from user. 
# Check if the student is eligible for admission: 
# Age must be between 18 and 25 AND marks >= 60 

age = int(input("Enter age: "))
marks = float(input("Enter marks: "))

if age >= 18 and age <= 25 and marks >= 60:
    print("Eligible for Admission")
else:
    print("Not Eligible for Admission")
```

```python
# 2. Login Check

username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin" and password == "12345":
    print("Login Successful")
else:
    print("Login Failed")
```

```python
# 3. between 50 and 100 and NOT divisible by 5

num = int(input("Enter a number: "))

if num >= 50 and num <= 100 and not (num % 5 == 0):
    print("Number is valid")
else:
    print("Number is not valid")
```

```python
# 4.  largest number using 'and' and 'or' operators. 

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if (a >= b and a >= c):
    print("Largest:", a)
elif (b >= a and b >= c):
    print("Largest:", b)
else:
    print("Largest:", c)
```

```python
# 5. string and string slicing

name = input("Enter your full name: ")

print("Uppercase:", name.upper())
print("Lowercase:", name.lower())
print("Length:", len(name))
print("First 3 characters:", name[:3])
print("Last 3 characters:", name[-3:])
```

```python
# 6. Sentence slicing

sentence = input("Enter a sentence: ")

print("First 10 characters:", sentence[:10])
print("Last 10 characters:", sentence[-10:])
print("Index 5 to 15:", sentence[5:16])
print("Every 2nd character:", sentence[::2])
```

```python
# 7. Palidrome

word = input("Enter a word: ")

if word == word[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
# 8. Email Username and Domain

email = input("Enter your email: ")

username = email.split("@")[0]
domain = email.split("@")[1]

print("Username:", username)
print("Domain:", domain)
```

```python
# 9. Strong Password

password = input("Enter password: ")

if len(password) >= 8 and ("@" in password or "#" in password):
    print("Strong Password")
else:
    print("Weak Password")
```

```python
# 10. Student Pass or Fail

english = int(input("Enter English marks: "))
math = int(input("Enter Math marks: "))
science = int(input("Enter Science marks: "))

if english >= 40 and math >= 40 and science >= 40:
    print("Pass")
else:
    print("Fail")
```

```python
# 11. Positive even or negative or odd

num = int(input("Enter a number: "))

if (num > 0 and num % 2 == 0):
    print("Positive and Even")
elif num < 0:
    print("Negative")
elif num % 2 != 0:
    print("Odd")
else:
    print("Zero")
```

```python
# 12. City name Operators

city = input("Enter your city name: ")

print("Reverse:", city[::-1])

vowels = "aeiouAEIOU"
count = 0

for ch in city:
    if ch in vowels:
        count += 1

print("Number of vowels:", count)
print("Without first and last character:", city[1:-1])
```

```python
# 13. Mobile number slicing

mobile = input("Enter 10 digit mobile number: ")

print("First 3 digits:", mobile[:3])
print("Last 4 digits:", mobile[-4:])
print("Middle 3 digits:", mobile[3:6])
```

```python
# 14. Valid String

text = input("Enter a string: ")

if len(text) > 5 and (text.startswith("A") or text.startswith("a")):
    print("Valid String")
else:
    print("Invalid String")
```
