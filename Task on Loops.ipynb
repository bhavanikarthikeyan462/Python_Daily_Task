# Task on Loops

```python
# 1. print numbers from 1 to 10

for i in range(1, 11):
    print(i)
```

```python
# 2. print even numbers from 1 to 20

for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
# 3. multiplication table using for loop

num = int(input("Enter a number: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
# 4. sum of first 10 natural numbers

total = 0

for i in range(1, 11):
    total = total + i

print("Sum:", total)
```

```python
# 5. print star pattern

for i in range(1, 6):
    print("*" * i)
```

```python
# 6. print square of 5 numbers

numbers = [2, 4, 6, 8, 10]

for num in numbers:
    print(num, "square =", num * num)
```

```python
# 7. print numbers from 50 down to 30

for i in range(50, 29, -1):
    print(i)
```

```python
# 8. count numbers divisible by 3 between 1 and 50

count = 0

for i in range(1, 51):
    if i % 3 == 0:
        count += 1

print("Count:", count)
```

```python
# 9. print each character of a string

text = input("Enter a string: ")

for ch in text:
    print(ch)
```

```python
# 10. multiplication tables from  1 to 5 

for num in range(1, 6):
    print("Table of", num)

    for i in range(1, 11):
        print(num, "x", i, "=", num * i)

    print()
```

```python
# 11. print numbers from 1 to 15

i = 1

while i <= 15:
    print(i)
    i += 1
```

```python
# 12. multiplication table using while

num = int(input("Enter a number: "))

i = 1

while i <= 10:
    print(num, "x", i, "=", num * i)
    i += 1
```

```python
# 13. sum numbers until users enters 0

total = 0

while True:
    num = int(input("Enter a number: "))

    if num == 0:
        break

    total += num

print("Sum:", total)
```

```python
# 14. find odd numbers from 1 to 20

i = 1

while i <= 20:
    if i % 2 != 0:
        print(i)

    i += 1
```

```python
# 15. keep dividing by 2 until less than 1

num = float(input("Enter a positive number: "))

steps = 0

while num >= 1:
    num = num / 2
    steps += 1

print("Number of steps:", steps)
```

```python
# 16. reverse a number using while

num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number:", reverse)
```

```python
# 17. keep taking input until stop

while True:
    text = input("Enter something: ")

    if text.lower() == "stop":
        print("Program stopped")
        break

    print("You entered:", text)
```

```python
# 18. fibonacci series-10 terms

a = 0
b = 1
count = 0

while count < 10:
    print(a, end=" ")

    next_number = a + b
    a = b
    b = next_number

    count += 1
```

```python
# 19. check whether a number is prime using while loop

num = int(input("Enter a number: "))

if num < 2:
    print("Not a Prime Number")
else:
    i = 2
    is_prime = True

    while i < num:
        if num % i == 0:
            is_prime = False
            break
        i += 1

    if is_prime:
        print("Prime Number")
    else:
        print("Not a Prime Number")
```

```python
# 20. menu- driven program using while loop

while True:
    print("\n--- MENU ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Addition:", a + b)

    elif choice == 2:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Subtraction:", a - b)

    elif choice == 3:
        print("Program Exited")
        break

    else:
        print("Invalid Choice")
```
