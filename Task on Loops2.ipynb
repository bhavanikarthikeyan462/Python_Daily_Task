# Task on Loops

```python
# 1. print numbers from 1 to 10

for i in range(1, 11):
    print(i, end=" ")
```

```python
# 2. sum of first n natural numbers

n = int(input("Enter n: "))

total = 0

for i in range(1, n + 1):
    total += i

print("Sum =", total)
```

```python
# 3. multiplication table

num = int(input("Enter a number: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
# 4. print even numbers from 1 to 20

for i in range(1, 21):
    if i % 2 == 0:
        print(i, end=" ")
```

```python
# 5. count numbers divisible by 3 between 1 to 50

count = 0

for i in range(1, 51):
    if i % 3 == 0:
        count += 1

print("Total numbers divisible by 3:", count)
```

```python
# 6. print characters of a string

text = input("Enter a string: ")

for ch in text:
    print(ch)
```

```python
# 7. print numbers from 50 to 30 in descending order

for i in range(50, 29, -1):
    print(i, end=" ")
```

```python
# 8. find factorial of a number

n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)
```

```python
# 9. print numbers from 1 to 15

i = 1

while i <= 15:
    print(i, end=" ")
    i += 1
```

```python
# 10. take numbers until 0 and print sum

total = 0

while True:
    num = int(input("Enter a number: "))

    if num == 0:
        break

    total += num

print("Sum =", total)
```

```python
# 11. reverse a number

num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

print("Reverse =", reverse)
```

```python
# 12. check whether a number is prime

n = int(input("Enter a number: "))

i = 2
prime = True

if n < 2:
    prime = False

while i * i <= n:
    if n % i == 0:
        prime = False
        break
    i += 1

if prime:
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
# 13. fibonnacci series using while number

n = int(input("Enter number of terms: "))

a = 0
b = 1
i = 0

while i < n:
    print(a, end=" ")
    a, b = b, a + b
    i += 1
```

```python
# 14. ask password untill correct

correct_password = "python123"

while True:
    password = input("Enter password: ")

    if password == correct_password:
        print("Access Granted")
        break
    else:
        print("Wrong Password")
```

```python
# 15. print odd numbers from 1 to 30

i = 1

while i <= 30:
    print(i, end=" ")
    i += 2
```

```python
# 16. sum of digits

num = int(input("Enter a number: "))

total = 0

while num > 0:
    digit = num % 10
    total += digit
    num //= 10

print("Sum of digits =", total)
```

```python
# 17. star pattern

for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# 18. multiplication table from 1 to 5

for i in range(1, 6):
    for j in range(1, 11):
        print(i * j, end=" ")
    print()
```

```python
# 19. print numbers from 1 to 100

for num in range(2, 101):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

```python
# 20. number pattern

for i in range(1, 5):
    for j in range(i):
        print(i, end=" ")
    print()
```

```python
# 21. sum from 1 to n using nested loop

n = int(input("Enter n: "))

total = 0

for i in range(1, n + 1):
    for j in range(i, i + 1):
        total += j

print("Sum =", total)
```

```python
# 22. hollow square pattern

n = int(input("Enter size: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# 23. floyd's triangle

n = int(input("Enter number of rows: "))

num = 1

for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
# 24. stop when number 13 appears - break

for i in range(1, 21):
    if i == 13:
        break
    print(i, end=" ")
```

```python
# 25. prime numbers from 1 to 30 except 

for i in range(1, 31):
    if i % 5 == 0:
        continue
    print(i, end=" ")
```
