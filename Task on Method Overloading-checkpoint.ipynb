# Method Overloading

```python
# 1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method name.

class Calculator:

    def add(self, a, b, c = None):
        if c is None:
            return a + b
        else:
            return a + b + c

obj = Calculator()

print("Sum of 2 Numbers:", obj.add(10,20))
print("Sum of 3 Numbers:", obj.add(30,40))
```

```python
# 2. Create an Area class that calculates the area of a square and rectangle using the same
# method.

class Area:

    def calculate(self,length,breadth = None):
        if breadth is None:
            return length * length
        else:
            return length * breadth

obj = Area()

print("square area:", obj.calculate(10,5))
print("rectangle area:", obj.calculate(9,5))
        
```

```python
# 3. Create a Student class with a details() method that accepts different numbers of
# arguments.

class Student:

    def details(self, name, age = None, city = None):
        print("name:",name)

        if age is not None:
            print("Age:",age)

        if city is not None:
            print("City:",city)

obj = Student()

obj.details("Arsheya")
print()

obj.details("Arsheya",25)
print()

obj.details("Arsheya",25,"Hyderabad")
```

```python
# 4. Create a MathOperations class that multiplies any number of values.

class MathOperations:

    def multiply(self,*args):
        result = 1
        for i in args:
            result *= i
        return result

obj = MathOperations()

print(obj.multiply(2,3))
print(obj.multiply(2,3,4))
print(obj.multiply(2,3,4,5))
print(obj.multiply(2,3,4,5,6))
print(obj.multiply(2,3,4,5,6,7))
```

```python
# 5. Create a Display class that prints different messages based on the argument type.

class Display:

    def show(self,value):
        if isinstance(value,int):
           print("Integer:",value)
        elif isinstance(value,str):
            print("String:",value)
        elif isinstance(value,list):
            print("List:",value)
        else:
            print("Unknown Value")

obj = Display()

obj.show(100)
obj.show("Python")
obj.show([10,20,30])
```

```python
# 6. Create a Volume class that calculates the volume of a cube and cuboid.

class Volume:
    def Calculate(self,length,breadth = None,height = None):
        if breadth is None and height is None:
            return length ** 3
        else:
            return length * breadth * height

obj = Volume()
print("Cube Volume:",obj.Calculate(4))
print("Cubiod Volume:",obj.Calculate(4,5,6))
```

```python
# 7. Create a PrintData class that prints one or many values.

class PrintData:

    def print_data(self,*args):
        for i in args:
            print(i)

obj = PrintData()
obj.print_data("Hello")
print()

obj.print_data(10,20,30,"Python")
```

# Method Overriding

```python
# 8. Create an Animal class with sound(). Override it in Dog and Cat.

class Animal:

    def sound(self):
        print("Animal Makes a sound")

class Dog(Animal):

    def sound(self):
        print("Dog Barks")

class Cat(Animal):

    def sound(self):
        print("Cat Meows")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
# 10. Create an Employee class and override calculate_salary() in Manager.

class Employee:

    def __init__(self,salary):
        self.salary = salary

    def calculate_salary(self):
        return self.salary

class Manager(Employee):

    def __init__(self,salary,bonus):
        super().__init__(salary)
        self.bonus = bonus

    def calculate_salary(self):
        return self.salary + self.bonus

e = Employee(30000)
m = Manager(50000,100000)

print("Employee Salary:",e.calculate_salary())
print("Manager Salary:",m.calculate_salary())
```

```python
# 11. Create Shape, Circle, and Rectangle classes.

class Shape:

    def area(self):
        print("Area of Shape")

class Circle(Shape):

    def __init__(self,radius):
        self.radius = radius

    def area(self):
        print("Area of circle:",3.14 * self.radius * self.radius)
        
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        print("Area of Rectangle:",self.length * self.width)

c = Circle(5)
r = Rectangle(4,6)

c.area()
r.area()
```

```python
# 13. Create Parent and Child classes using super().

class Parent:
    def display(self):
        print("This is Parent Class")

class Child(Parent):
    def display(self):
        super().display()
        print("This is Child Class")

obj = Child()
obj.display()
```

# Operator Overloading

```python
# 15. Create a Point class and overload the + operator.

class Point:

    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self,other):
        return Point(self.x + other.x,self.y + other.y)

    def display(self):
        print("Point(", self.x, ",", self.y,")")

p1 = Point(2,3)
p2 = Point(4,5)

p3 = p1 + p2
p3.display()
```

```python
# 16. Create a Book class and overload the + operator.

class Book:
    def __init__(self,pages):
        self.pages = pages

    def __add__(self,other):
        return self.pages + other.pages

b1 = Book(200)
b2 = Book(300)

print("Total Pages:", b1 + b2)
```

```python
# 17. Create a Student class and overload the > operator.

class Student:
    def __init__(self,marks):
        self.marks = marks

    def __gt__(self,other):
        return self.marks > other.marks

s1 = Student(90)
s2 = Student(80)

print(s1 > s2)
```

```python
# 18. Create a Rectangle class and overload the == operator.

class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def __eq__(self,other):
        return self.length * self.width == other.length * other.width

r1 = Rectangle(4,5)
r2 = Rectangle(2,10)

print(r1 == r2)
```

```python
# 19. Create a Money class and overload the + and - operators.

class Money:
    def __init__(self,amount):
        self.amount = amount

    def __add__(self,other):
        return Money(self.amount + other.amount)

    def __sub__(self,other):
        return Money(self.amount - other.amount)

    def display(self):
        print("Amount:",self.amount)

m1 = Money(1000)
m2 = Money(300)

add = m1 + m2
sub = m1 - m2

add.display()
sub.display()
```

```python
# 20. Create a Distance class and overload <, >, and ==.

class Distance:
    def __init__(self,value):
        self.value = value

    def __it__(self,other):
        return self.value < other.value

    def __gt__(self,other):
        return self.value > other.value

    def __eq__(self,other):
        return self.value == other.value

d1 = Distance(15)
d2 = Distance(20)

print(d1 < d2)
print(d1 > d2)
print(d1 == d2)
```
