# 1–10: Basic Multiple Inheritance

```python
# 1. Father + Mother → Child

class Father:
    def father_method(self):
        print("Father's method")


class Mother:
    def mother_method(self):
        print("Mother's method")


class Child(Father, Mother):
    pass


c = Child()
c.father_method()
c.mother_method()
```

```python
# 2. A + B → C

class A:
    def show(self):
        print("Show method from A")


class B:
    def display(self):
        print("Display method from B")


class C(A, B):
    pass


obj = C()
obj.show()
obj.display()
```

```python
# 3. Teacher + Author → Person

class Teacher:
    def __init__(self, subject):
        self.subject = subject


class Author:
    def __init__(self, book):
        self.book = book


class Person(Teacher, Author):
    def __init__(self, subject, book, name):
        Teacher.__init__(self, subject)
        Author.__init__(self, book)
        self.name = name

    def display(self):
        print("Name:", self.name)
        print("Subject:", self.subject)
        print("Book:", self.book)


p = Person("Python", "Python Programming", "Rahul")
p.display()
```

```python
# 4. Sports + Academic → Student

class Sports:
    def play(self):
        print("Student plays cricket")


class Academic:
    def study(self):
        print("Student studies Python")


class Student(Sports, Academic):
    pass


s = Student()
s.play()
s.study()
```

```python
# 5. Camera + MusicPlayer → Smartphone

class Camera:
    def take_photo(self):
        print("Photo taken")


class MusicPlayer:
    def play_music(self):
        print("Playing music")


class Smartphone(Camera, MusicPlayer):
    pass


phone = Smartphone()
phone.take_photo()
phone.play_music()
```

```python
# 6. Same method name — resolve conflict

class A:
    def show(self):
        print("Show from A")


class B:
    def show(self):
        print("Show from B")


class C(A, B):
    def show(self):
        print("Show from C")
        A.show(self)
        B.show(self)


obj = C()
obj.show()
```

```python
# 7. Engine + Wheels → Car

class Engine:
    def start_engine(self):
        print("Engine started")


class Wheels:
    def rotate_wheels(self):
        print("Wheels are rotating")


class Car(Engine, Wheels):
    pass


car = Car()
car.start_engine()
car.rotate_wheels()
```

```python
# 10. Demonstrate MRO

class A:
    def show(self):
        print("A")


class B(A):
    def show(self):
        print("B")


class C(A):
    def show(self):
        print("C")


class D(B, C):
    pass


obj = D()

obj.show()

print(D.mro())
```

```python
# 11–18: Multiple Inheritance with Static Methods


# 11. Two parents with static methods

class A:
    @staticmethod
    def method_a():
        print("Static method from A")


class B:
    @staticmethod
    def method_b():
        print("Static method from B")


class C(A, B):
    pass


obj = C()
obj.method_a()
obj.method_b()
```

```python
# 12. MathUtils → AdvancedMath

class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b


class AdvancedMath(MathUtils):
    def square(self, n):
        return n * n


obj = AdvancedMath()

print(obj.add(10, 20))
print(obj.square(5))
```

```python
# 13. Validator → Form

class Validator:
    @staticmethod
    def is_valid(value):
        return value != ""


class Form(Validator):
    def check(self, value):
        if self.is_valid(value):
            print("Valid")
        else:
            print("Invalid")


form = Form()
form.check("Hello")
```

```python
# 14. Converter + another class → WeatherApp

class Converter:
    @staticmethod
    def to_celsius(fahrenheit):
        return (fahrenheit - 32) * 5 / 9


class Weather:
    def show_weather(self):
        print("Today's weather is sunny")


class WeatherApp(Converter, Weather):
    pass


app = WeatherApp()

print(app.to_celsius(98.6))
app.show_weather()
```

```python
# 15. Logger + Application

class Logger:
    @staticmethod
    def log(message):
        print("LOG:", message)


class Database:
    def connect(self):
        print("Database connected")


class Application(Logger, Database):
    pass


app = Application()

app.log("Application started")
app.connect()
```

```python
# 16. Same static method name

class A:
    @staticmethod
    def display():
        print("Display from A")


class B:
    @staticmethod
    def display():
        print("Display from B")


class C(A, B):
    @staticmethod
    def display():
        print("Display from C")


C.display()
```

```python
# 17. Calculator + Display

class Calculator:
    @staticmethod
    def add(a, b):
        return a + b


class Display:
    def show(self, value):
        print("Result:", value)


class MyCalculator(Calculator, Display):
    pass


obj = MyCalculator()

result = obj.add(10, 20)
obj.show(result)
```

```python
# 18. Calling static method using parent and child class

class A:
    @staticmethod
    def message():
        print("Hello from A")


class B:
    @staticmethod
    def welcome():
        print("Welcome from B")


class C(A, B):
    pass


# Using parent class
A.message()
B.welcome()

# Using child class
C.message()
C.welcome()
```

```python
# 19–25: Multiple Inheritance with Abstract Methods


# 19. Shape + Color

from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Color(ABC):
    @abstractmethod
    def get_color(self):
        pass


class Circle(Shape, Color):
    def area(self):
        return 3.14 * 5 * 5

    def get_color(self):
        return "Red"


obj = Circle()

print("Area:", obj.area())
print("Color:", obj.get_color())
```

```python
# 20. Payment + Notification

from abc import ABC, abstractmethod


class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class Notification(ABC):
    @abstractmethod
    def notify(self):
        pass


class OnlinePayment(Payment, Notification):
    def pay(self, amount):
        print("Paid:", amount)

    def notify(self):
        print("Payment notification sent")


obj = OnlinePayment()
obj.pay(500)
obj.notify()
```

```python
# 21. Vehicle + Electric → ElectricCar

from abc import ABC, abstractmethod


class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass


class Electric(ABC):
    @abstractmethod
    def charge(self):
        pass


class ElectricCar(Vehicle, Electric):
    def start(self):
        print("Electric car started")

    def charge(self):
        print("Car is charging")


car = ElectricCar()
car.start()
car.charge()
```

```python
# 22. Readable + Writable → File

from abc import ABC, abstractmethod


class Readable(ABC):
    @abstractmethod
    def read(self):
        pass


class Writable(ABC):
    @abstractmethod
    def write(self, data):
        pass


class File(Readable, Writable):
    def read(self):
        print("Reading file")

    def write(self, data):
        print("Writing:", data)


file = File()
file.read()
file.write("Hello Python")
```

```python
# 23. Flyable + Swimmable → Duck

from abc import ABC, abstractmethod


class Flyable(ABC):
    @abstractmethod
    def fly(self):
        pass


class Swimmable(ABC):
    @abstractmethod
    def swim(self):
        pass


class Duck(Flyable, Swimmable):
    def fly(self):
        print("Duck is flying")

    def swim(self):
        print("Duck is swimming")


duck = Duck()
duck.fly()
duck.swim()
```

```python
# 24. One abstract + one normal class

from abc import ABC, abstractmethod


class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass


class Horn:
    def horn(self):
        print("Horn is blowing")


class Car(Vehicle, Horn):
    def start(self):
        print("Car started")


car = Car()
car.start()
car.horn()
```

```python
# 25. Abstract methods with same name

from abc import ABC, abstractmethod


class A(ABC):
    @abstractmethod
    def show(self):
        pass


class B(ABC):
    @abstractmethod
    def show(self):
        pass


class C(A, B):
    def show(self):
        print("Single implementation resolves both abstract methods")


obj = C()
obj.show()
```

```python
# 26–30: Problem / Hard Level


# 26. Employee + Person → Manager

class Employee:
    def get_salary(self):
        return 50000


class Person:
    def get_details(self):
        return "Name: Rahul"


class Manager(Employee, Person):
    def get_manager_details(self):
        details = self.get_details()
        salary = self.get_salary()

        print(details)
        print("Salary:", salary)


manager = Manager()
manager.get_manager_details()
```

```python
# 27. Device + Network → SmartPhone

from abc import ABC, abstractmethod


class Device(ABC):
    @abstractmethod
    def power_on(self):
        pass


class Network:
    def connect(self):
        print("Connected to network")


class SmartPhone(Device, Network):
    def power_on(self):
        print("Smartphone powered on")


phone = SmartPhone()

phone.power_on()
phone.connect()
```

```python
# 28. Writer + Editor → Author with static counter

class Writer:
    def write(self):
        print("Author is writing")


class Editor:
    def edit(self):
        print("Author is editing")


class Author(Writer, Editor):
    total_authors = 0

    def __init__(self, name):
        self.name = name
        Author.total_authors += 1

    @staticmethod
    def count_authors():
        return Author.total_authors


a1 = Author("Rahul")
a2 = Author("Amit")
a3 = Author("Priya")

a1.write()
a2.edit()

print("Total authors:", Author.count_authors())
```

```python
# 29. Printable + Scannable → PrinterScanner

from abc import ABC, abstractmethod


class Printable(ABC):
    @abstractmethod
    def print_document(self):
        pass


class Scannable(ABC):
    @abstractmethod
    def scan_document(self):
        pass


class PrinterScanner(Printable, Scannable):
    def print_document(self):
        print("Printing document")

    def scan_document(self):
        print("Scanning document")


device = PrinterScanner()

device.print_document()
device.scan_document()
```

```python
# 30. Static method + Abstract method

from abc import ABC, abstractmethod


class A:
    @staticmethod
    def message():
        print("Static method from A")


class B(ABC):
    @abstractmethod
    def show(self):
        pass


class C(A, B):
    def show(self):
        print("Abstract method implemented in C")


obj = C()

obj.show()
obj.message()

# Static method can also be called using class name
C.message()
```
