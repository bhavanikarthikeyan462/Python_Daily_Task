```python
# 4. largest of three numbers using nested if

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a > b:
    if a > c:
        print("Largest:", a)
    else:
        print("Largest:", c)
else:
    if b > c:
        print("Largest:", b)
    else:
        print("Largest:", c)
```

```python
# 8. valid triangle and traingle types

a = float(input("Enter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    print("Valid Triangle")

    if a == b:
        if b == c:
            print("Equilateral")
        else:
            print("Isosceles")
    else:
        if b == c:
            print("Isosceles")
        elif a == c:
            print("Isosceles")
        else:
            print("Scalene")
else:
    print("Invalid Triangle")
```

```python
# 11. leap year using nested conditions

year = int(input("Enter year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("Leap Year")
        else:
            print("Not a Leap Year")
    else:
        print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
# 13. check equal numbers using nested if

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a == b:
    if b == c:
        print("All three numbers are equal")
    else:
        print("First and second numbers are equal")
else:
    if a == c:
        print("First and third numbers are equal")
    elif b == c:
        print("Second and third numbers are equal")
    else:
        print("All three numbers are different")
```

```python
# 18. valid triangle and right angled triangle

a = float(input("Enter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    print("Valid Triangle")

    if a * a + b * b == c * c:
        print("Right-Angled Triangle")
    elif a * a + c * c == b * b:
        print("Right-Angled Triangle")
    elif b * b + c * c == a * a:
        print("Right-Angled Triangle")
    else:
        print("Not a Right-Angled Triangle")
else:
    print("Invalid Triangle")
```
