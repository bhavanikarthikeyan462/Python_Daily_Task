```python
# 11. BankAccount with custom exception

class NegativeBalanceError(Exception):
    pass


class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance = self.__balance + amount

    def withdraw(self, amount):
        if self.__balance - amount < 0:
            raise NegativeBalanceError("Balance cannot become negative")
        self.__balance = self.__balance - amount

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)

account.deposit(500)
print("Balance:", account.get_balance())

account.withdraw(300)
print("Balance:", account.get_balance())
```

```python
# 15. Person → Student

class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)


s = Student("Rahul", 90)
s.display()
```

```python
# 16. Person → Employee → Manager

class Person:
    def show(self):
        print("I am a person")


class Employee(Person):
    def show(self):
        super().show()
        print("I am an employee")


class Manager(Employee):
    def show(self):
        super().show()
        print("I am a manager")


m = Manager()
m.show()
```

```python
# 17. Hierarchical inheritance

class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog says Woof")


class Cat(Animal):
    def sound(self):
        print("Cat says Meow")


d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
# 18. Abstract Shape, Circle and Square

from abc import ABC, abstractmethod


class Shape(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


class Square(Shape):
    def __init__(self, side):
        super().__init__("Square")
        self.side = side

    def area(self):
        return self.side * self.side


c = Circle(5)
s = Square(4)

print(c.name, c.area())
print(s.name, s.area())
```

```python
# 19. Multiple inheritance

class Father:
    def show(self):
        print("Father method")


class Mother:
    def show(self):
        print("Mother method")


class Child(Father, Mother):
    def show(self):
        Father.show(self)
        Mother.show(self)
        print("Child method")


c = Child()
c.show()
```

```python
# 21. Point and operator overloading

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __gt__(self, other):
        distance1 = self.x ** 2 + self.y ** 2
        distance2 = other.x ** 2 + other.y ** 2

        return distance1 > distance2

    def display(self):
        print("(", self.x, ",", self.y, ")")


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2
p3.display()

print(p1 > p2)
```

```python
# 26. Student with marks list

class EmptyMarksError(Exception):
    pass


class Student:
    def __init__(self, marks):
        self.__marks = []
        self.set_marks(marks)

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if len(marks) == 0:
            raise EmptyMarksError("Marks list cannot be empty")

        for mark in marks:
            if mark < 0 or mark > 100:
                raise ValueError("Marks must be between 0 and 100")

        self.__marks = marks

    def average(self):
        return sum(self.__marks) / len(self.__marks)

    def __gt__(self, other):
        return self.average() > other.average()


s1 = Student([80, 90, 85])
s2 = Student([70, 75, 80])

print("Student 1 average:", s1.average())
print("Student 2 average:", s2.average())

print("Student 1 has higher average:", s1 > s2)
```

```python
# 27. Menu-driven banking system with OOP + File Handling

class InsufficientBalanceError(Exception):
    pass


class BankAccount:
    def __init__(self, account_number, name, balance):
        self.account_number = account_number
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Invalid deposit amount")

        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Invalid withdrawal amount")

        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance")

        self.balance = self.balance - amount

    def save(self):
        file = open("accounts.txt", "a")
        file.write(
            self.account_number + "," +
            self.name + "," +
            str(self.balance) + "\n"
        )
        file.close()


try:
    while True:
        print("\n1. Create Account")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Display File")
        print("5. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            account_number = input("Account number: ")
            name = input("Name: ")
            balance = float(input("Initial balance: "))

            account = BankAccount(account_number, name, balance)
            account.save()

            print("Account saved")

        elif choice == "2":
            amount = float(input("Deposit amount: "))
            print("Deposit operation completed")

        elif choice == "3":
            amount = float(input("Withdrawal amount: "))
            print("Withdrawal operation completed")

        elif choice == "4":
            try:
                file = open("accounts.txt", "r")
                print(file.read())
                file.close()

            except FileNotFoundError:
                print("Account file not found")

        elif choice == "5":
            break

        else:
            print("Invalid choice")

except ValueError:
    print("Please enter a valid value")

except InsufficientBalanceError as e:
    print(e)
```

```python
# 28. Student management system

class StudentNotFoundError(Exception):
    pass


students = {}


def add_student():
    name = input("Enter student name: ")
    marks = float(input("Enter marks: "))

    students[name] = marks
    print("Student added")


def update_student():
    name = input("Enter student name: ")

    if name not in students:
        raise StudentNotFoundError("Student not found")

    marks = float(input("Enter new marks: "))
    students[name] = marks

    print("Marks updated")


def delete_student():
    name = input("Enter student name: ")

    if name not in students:
        raise StudentNotFoundError("Student not found")

    del students[name]
    print("Student deleted")


def display_students():
    for name, marks in students.items():
        print(name, marks)


while True:
    print("\n1. Add")
    print("2. Update")
    print("3. Delete")
    print("4. Display")
    print("5. Exit")

    choice = input("Enter choice: ")

    try:
        if choice == "1":
            add_student()

        elif choice == "2":
            update_student()

        elif choice == "3":
            delete_student()

        elif choice == "4":
            display_students()

        elif choice == "5":
            break

        else:
            print("Invalid choice")

    except StudentNotFoundError as e:
        print(e)

    except ValueError:
        print("Please enter valid marks")
```

```python
# 29. Employee salary, history, tax and comparison

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
        self.history = [salary]

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):
        increase = new_salary - self.__salary
        percentage = (increase / self.__salary) * 100

        if percentage > 15:
            print("Salary cannot increase by more than 15%")
        else:
            self.__salary = new_salary
            self.history.append(new_salary)

    @staticmethod
    def calculate_tax(salary):
        return salary * 0.10

    def __gt__(self, other):
        return self.__salary > other.__salary


e1 = Employee("Rahul", 50000)
e2 = Employee("Priya", 60000)

e1.set_salary(55000)

print("Rahul salary:", e1.get_salary())
print("Salary history:", e1.history)

print("Tax:", Employee.calculate_tax(e1.get_salary()))

print("Rahul earns more:", e1 > e2)
```

```python
# 30. Complete Vehicle System
# Abstract class
# Multiple inheritance
# Operator overloading
# Private attributes
# Getter/setter
# Custom exception
# File handling
# Duck typing

from abc import ABC, abstractmethod


# Custom exception
class InvalidSpeedError(Exception):
    pass


# Abstract class
class Vehicle(ABC):
    def __init__(self, brand, speed):
        self.brand = brand
        self.__speed = 0
        self.set_speed(speed)

    def get_speed(self):
        return self.__speed

    def set_speed(self, speed):
        if speed < 0:
            raise InvalidSpeedError("Speed cannot be negative")

        self.__speed = speed

    @abstractmethod
    def start(self):
        pass

    def __gt__(self, other):
        return self.__speed > other.__speed


# First parent class
class Electric:
    def charge(self):
        print("Vehicle is charging")


# Second parent class
class Fuel:
    def refuel(self):
        print("Vehicle is being refueled")


# Multiple inheritance
class ElectricCar(Vehicle, Electric):
    def start(self):
        print(self.brand, "electric car started")


class FuelCar(Vehicle, Fuel):
    def start(self):
        print(self.brand, "fuel car started")


# Create objects
car1 = ElectricCar("Tesla", 180)
car2 = FuelCar("Toyota", 160)

# Getter
print("Car 1 speed:", car1.get_speed())
print("Car 2 speed:", car2.get_speed())

# Setter
car1.set_speed(190)
print("Updated Car 1 speed:", car1.get_speed())

# Duck typing
car1.start()
car2.start()

# Multiple inheritance methods
car1.charge()
car2.refuel()

# Operator overloading
print("Car 1 is faster:", car1 > car2)


# File handling
try:
    file = open("vehicles.txt", "w")

    file.write(
        car1.brand + "," +
        str(car1.get_speed()) + "\n"
    )

    file.write(
        car2.brand + "," +
        str(car2.get_speed()) + "\n"
    )

    file.close()

    print("Vehicle details saved")


except InvalidSpeedError as e:
    print(e)


# Read file
try:
    file = open("vehicles.txt", "r")

    print("\nVehicle Details:")
    print(file.read())

    file.close()

except FileNotFoundError:
    print("Vehicle file not found")
```
