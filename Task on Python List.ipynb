# Task on python list

```python
# 1. create a list of 5 students

students = ["Rahul", "Priya", "Arun", "Sneha", "Kiran"]

print(students)
```

```python
# 2. print type and length of a list

numbers = [10, 20, 30, 40, 50]

print("Type:", type(numbers))
print("Length:", len(numbers))
```

```python
# 3. append 3 numbers to an empty list

numbers = []

numbers.append(10)
numbers.append(20)
numbers.append(30)

print(numbers)
```

```python
# 4. use extend() to add elements

lst = [1, 2, 3]

lst.extend([4, 5, 6])

print(lst)
```

```python
# 5. insert 100 at index 2

numbers = [10, 20, 30, 40]

numbers.insert(2, 100)

print(numbers)
```

```python
# 6. remove element 50

numbers = [10, 20, 50, 30, 50]

numbers.remove(50)

print(numbers)
```

```python
# 7. remove the last element using pop()

lst = [5, 10, 15, 20]

lst.pop()

print(lst)
```

```python
# 8. clear all elements

data = [1, 2, 3, 4, 5]

data.clear()

print(data)
```

```python
# 9. find the index of 25

numbers = [10, 15, 25, 30, 40]

index = numbers.index(25)

print("Index of 25:", index)
```

```python
# 10. count how many times 7 appears

numbers = [7, 2, 7, 5, 7, 8]

count = numbers.count(7)

print("7 appears", count, "times")
```

```python
# 11. sort in ascending order

numbers = [45, 12, 78, 23, 56]

numbers.sort()

print(numbers)
```

```python
# 12. sort in descending order

numbers = [45, 12, 78, 23, 56]

numbers.sort(reverse=True)

print(numbers)
```

```python
# 13. reverse a list using reverse()

fruits = ["apple", "banana", "mango", "orange"]

fruits.reverse()

print(fruits)
```

```python
# 14. create a copy using copy()

original = [1, 2, 3, 4]

copy_list = original.copy()

print("Original:", original)
print("Copy:", copy_list)
```

```python
# 15. convert a sentence into a list of words

sentence = input("Enter a sentence: ")

words = sentence.split()

print("List of words:", words)
```

```python
# 16. append 3 numbers and sort descending

numbers = [10, 25, 5, 40, 15]

numbers.append(30)
numbers.append(50)
numbers.append(20)

numbers.sort(reverse=True)

print("List:", numbers)
```

```python
# 17. insert python,pop last element and find length

data = ["Java", "C", "SQL", "HTML"]

data.insert(0, "Python")

data.pop()

print("List:", data)
print("Length:", len(data))
```

```python
# 18. take 5 city names and print in reverse order

cities = []

for i in range(5):
    city = input("Enter city: ")
    cities.append(city)

cities.reverse()

print("Cities in reverse order:")
print(cities)
```

```python
# 19. remove all occurrences of 20

data = [10, 20, 30, 20, 40, 20]

while 20 in data:
    data.remove(20)

print("Updated list:", data)
```

```python
# 20. sort marks and find highest and lowest

marks = [78, 45, 92, 65, 88, 55]

marks.sort()

lowest = marks[0]
highest = marks[-1]

print("Sorted marks:", marks)
print("Lowest mark:", lowest)
print("Highest mark:", highest)
```
