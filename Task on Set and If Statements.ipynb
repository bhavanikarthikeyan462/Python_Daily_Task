```python
# Set Questions
# 1. Create a set with duplicate numbers

numbers = {1, 2, 3, 2, 4, 1}

print(numbers)
```

```python
# 2.Create an empty set and add 3 numbers

numbers = set()

numbers.add(10)
numbers.add(20)
numbers.add(30)

print(numbers)
```

```python
# 3. Add multiple elements using update()

s = {10, 20, 30}

s.update([40, 50, 60])

print(s)
```

```python
# 4. remove() vs discard()

s = {10, 20, 30, 40}

s.remove(20)
print("After remove:", s)

s.discard(30)
print("After discard:", s)
```

```python
# 5. Union, intersection, and difference

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("Difference:", A.difference(B))
```

```python
# 6. Check whether 25 exists

s = {10, 20, 30, 40}

if 25 in s:
    print("25 is present")
else:
    print("25 is not present")
```

```python
# 7. Create set from a list and find length

numbers = [1, 2, 2, 3, 4, 4, 5]

s = set(numbers)

print("Set:", s)
print("Length:", len(s))
```

```python
# 8. Take 5 numbers and store unique numbers

numbers = set()

for i in range(5):
    num = int(input("Enter number: "))
    numbers.add(num)

print("Unique numbers:", numbers)
```

```python
# 23. valid triangle and triangle type

a = float(input("Enter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    print("Valid Triangle")

    if a == b == c:
        print("Equilateral Triangle")
    elif a == b or b == c or a == c:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")
else:
    print("Invalid Triangle")
```

```python
# 24. calculate percentage and grade

m1 = float(input("Enter marks for subject 1: "))
m2 = float(input("Enter marks for subject 2: "))
m3 = float(input("Enter marks for subject 3: "))
m4 = float(input("Enter marks for subject 4: "))
m5 = float(input("Enter marks for subject 5: "))

total = m1 + m2 + m3 + m4 + m5
percentage = total / 5

print("Total Marks:", total)
print("Percentage:", percentage)

if percentage >= 90:
    print("Grade: A+")
elif percentage >= 80:
    print("Grade: A")
elif percentage >= 70:
    print("Grade: B")
elif percentage >= 60:
    print("Grade: C")
else:
    print("Grade: Fail")

if percentage >= 75:
    print("Passed with Distinction")
```
