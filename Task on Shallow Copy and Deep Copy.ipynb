# Task on Shallow Copy and Deep Copy

```python
# 1. Shallow copy of a nested list

import copy

original = [1, 2, [3, 4], 5]

shallow = copy.copy(original)

shallow[2][0] = 99

print("Original:", original)
print("Shallow:", shallow)
```

```python
# 2. Deep copy of a nested list

import copy

original = [1, 2, [3, 4], 5]

deep = copy.deepcopy(original)

deep[2][0] = 99

print("Original:", original)
print("Deep:", deep)
```

```python
# 3. Demonstrate shallow copy using copy.copy()

import copy

numbers = [10, 20, [30, 40]]

shallow = copy.copy(numbers)

print("Before change:")
print("Original:", numbers)
print("Shallow:", shallow)

shallow[2][0] = 100

print("\nAfter change:")
print("Original:", numbers)
print("Shallow:", shallow)
```

```python
# 4. Demonstrate deep copy using copy.deepcopy()

import copy

numbers = [10, 20, [30, 40]]

deep = copy.deepcopy(numbers)

print("Before change:")
print("Original:", numbers)
print("Deep:", deep)

deep[2][0] = 100

print("\nAfter change:")
print("Original:", numbers)
print("Deep:", deep)
```

```python
# 7. Manual deep copy without deepcopy()

def manual_deep_copy(lst):
    result = []

    for item in lst:
        if isinstance(item, list):
            result.append(manual_deep_copy(item))
        else:
            result.append(item)

    return result


original = [1, [2, 3], [4, [5, 6]]]

copy_list = manual_deep_copy(original)

copy_list[1][0] = 99

print("Original:", original)
print("Copy:", copy_list)
```

```python
# 8. Shallow and deep copies

import copy

data = [10, [20, 30], {"name": "Ram"}, 40]

shallow = copy.copy(data)
deep = copy.deepcopy(data)

data[1][0] = 99
data[2]["name"] = "Shyam"

print("Original:", data)
print("Shallow:", shallow)
print("Deep:", deep)
```

```python
# 9. Detect whether a copy is shallow or deep

def check_copy(original, copied):
    original[1][0] = 999

    if copied[1][0] == 999:
        print("This is a shallow copy.")
    else:
        print("This is a deep copy.")


import copy

original = [10, [20, 30]]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

check_copy(original, shallow)

original = [10, [20, 30]]

check_copy(original, deep)
```

```python
# 19. ATM

class ATM:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance = self.balance + amount
            print("Deposited:", amount)
            print("Balance:", self.balance)
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance = self.balance - amount
            print("Withdrawn:", amount)
            print("Balance:", self.balance)


atm = ATM(10000)

atm.deposit(5000)
atm.withdraw(3000)
atm.withdraw(0)
atm.withdraw(-500)
atm.withdraw(20000)
```

```python
# 23. Student Attendance

class StudentAttendance:
    def __init__(self, total_classes, attended_classes):
        self.total_classes = total_classes
        self.attended_classes = attended_classes

    def mark_attendance(self, number):
        if self.attended_classes + number <= self.total_classes:
            self.attended_classes = self.attended_classes + number

            percentage = (
                self.attended_classes / self.total_classes
            ) * 100

            print("Attendance marked:", number)
            print("Attendance percentage:", percentage, "%")
        else:
            print("Attendance cannot exceed total classes")


student = StudentAttendance(100, 70)

student.mark_attendance(10)
student.mark_attendance(30)
```

```python
# 24. Bus Reservation

class BusReservation:
    def __init__(self, bus_name, total_seats, booked_seats):
        self.bus_name = bus_name
        self.total_seats = total_seats
        self.booked_seats = booked_seats

    def reserve(self, number):
        if self.booked_seats + number <= self.total_seats:
            self.booked_seats = self.booked_seats + number

            remaining = self.total_seats - self.booked_seats

            print("Reservation successful")
            print("Remaining seats:", remaining)
        else:
            print("Not enough seats available")


bus = BusReservation("Volvo Express", 50, 40)

bus.reserve(5)
bus.reserve(10)
```

```python
# 25. Online Shopping Cart

class ShoppingCart:
    def __init__(self, customer_name, cart_total, wallet_balance):
        self.customer_name = customer_name
        self.cart_total = cart_total
        self.wallet_balance = wallet_balance

    def add_item(self, price):
        if price > 0:
            self.cart_total = self.cart_total + price
            print("Item added")
            print("Cart total:", self.cart_total)
        else:
            print("Invalid price")

    def remove_item(self, price):
        if price <= self.cart_total:
            self.cart_total = self.cart_total - price
            print("Item removed")
            print("Cart total:", self.cart_total)
        else:
            print("Cart total cannot become negative")

    def checkout(self):
        if self.cart_total == 0:
            print("Cart is empty")
        elif self.wallet_balance >= self.cart_total:
            self.wallet_balance = self.wallet_balance - self.cart_total
            self.cart_total = 0

            print("Payment successful")
            print("Remaining wallet balance:", self.wallet_balance)
            print("Cart total:", self.cart_total)
        else:
            print("Insufficient wallet balance")


cart = ShoppingCart("Ram", 0, 5000)

cart.add_item(1000)
cart.add_item(1500)

cart.remove_item(500)

cart.checkout()
```

```python
# Bonus Challenge — University Student Management System

class UniversityStudent:

    def __init__(self, student_id, name, age, course,
                 total_fee, paid_fee, marks,
                 total_classes, attended_classes):

        self.student_id = student_id
        self.name = name
        self.age = age
        self.course = course
        self.total_fee = total_fee
        self.paid_fee = paid_fee
        self.marks = marks
        self.total_classes = total_classes
        self.attended_classes = attended_classes

    # Update marks
    def update_marks(self, new_marks):
        if 0 <= new_marks <= 100:
            self.marks = new_marks
            print("Marks updated successfully")
        else:
            print("Invalid marks")

    # Pay fees
    def pay_fees(self, amount):
        remaining_fee = self.total_fee - self.paid_fee

        if amount <= remaining_fee:
            self.paid_fee = self.paid_fee + amount
            print("Fee payment successful")
        else:
            print("Payment exceeds remaining fee")

    # Update attendance
    def update_attendance(self, number):
        if self.attended_classes + number <= self.total_classes:
            self.attended_classes = self.attended_classes + number
            print("Attendance updated")
        else:
            print("Attendance cannot exceed total classes")

    # Change course
    def change_course(self, new_course):
        self.course = new_course
        print("Course changed to:", self.course)

    # Calculate attendance percentage
    def attendance_percentage(self):
        percentage = (
            self.attended_classes / self.total_classes
        ) * 100

        return percentage

    # Check scholarship
    def scholarship(self):
        attendance = self.attendance_percentage()

        if (
            self.marks > 90
            and attendance > 95
            and self.paid_fee == self.total_fee
        ):
            return "Yes"
        else:
            return "No"

    # Check pass/fail
    def result(self):
        if self.marks >= 35:
            return "Pass"
        else:
            return "Fail"

    # Display complete report
    def display_report(self):

        print("\n========== STUDENT REPORT CARD ==========")

        print("Student ID:", self.student_id)
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)
        print("Marks:", self.marks)
        print("Result:", self.result())

        print("\n---------- Fee Status ----------")

        print("Total Fee:", self.total_fee)
        print("Paid Fee:", self.paid_fee)
        print("Remaining Fee:", self.total_fee - self.paid_fee)

        if self.paid_fee == self.total_fee:
            print("Fee Status: Fully Paid")
        else:
            print("Fee Status: Pending")

        print("\n---------- Attendance ----------")

        print(
            "Attendance:",
            self.attendance_percentage(),
            "%"
        )

        print("\n---------- Scholarship ----------")

        print(
            "Scholarship Eligible:",
            self.scholarship()
        )

        print("==========================================")


# Create student object

student1 = UniversityStudent(
    101,
    "Ram",
    20,
    "Computer Science",
    100000,
    100000,
    95,
    100,
    98
)


# Update marks
student1.update_marks(96)

# Pay remaining fee
student1.pay_fees(1000)

# Update attendance
student1.update_attendance(1)

# Change course
student1.change_course("Artificial Intelligence")

# Display report
student1.display_report()
```
