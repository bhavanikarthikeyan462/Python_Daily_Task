```python
# 7. strip each word and join them

words = [" Python ", " is ", " awesome "]

cleaned_words = []

for word in words:
    cleaned_words.append(word.strip())

result = " ".join(cleaned_words)

print(result)
```

```python
# 14. replace bad with good after splitting and joining

sentence = input("Enter a sentence: ")

words = sentence.split()

for i in range(len(words)):
    if words[i] == "bad":
        words[i] = "good"

result = " ".join(words)

print("Result:", result)
```

```python
# 15. join numbers with-

numbers = ["10", "20", "30", "40"]

result = "-".join(numbers)

print(result)
```

```python
# 16. remove words starting with vowels

sentence = input("Enter a sentence: ")

words = sentence.split()
result_words = []

for word in words:
    if word[0].lower() not in "aeiou":
        result_words.append(word)

result = " ".join(result_words)

print("Result:", result)
```

```python
# 17. replace spaces with -, split, and print words with length > 4

text = input("Enter a string: ")

text = text.replace(" ", "-")

words = text.split("-")

print("Words with length > 4:")

for word in words:
    if len(word) > 4:
        print(word)
```

```python
# 18. replace the first student name

students = ["Rahul", "Priya", "Arun", "Sneha", "Kiran"]

result = ", ".join(students)

print("Original:", result)

result = result.replace(students[0], "New Student", 1)

print("Updated:", result)
```

```python
# 19. slit paragraph int sentsnces and join with \n-

paragraph = input("Enter a paragraph: ")

sentences = paragraph.split(".")

cleaned_sentences = []

for sentence in sentences:
    sentence = sentence.strip()
    
    if sentence:
        cleaned_sentences.append(sentence)

result = "\n- ".join(cleaned_sentences)

print("- " + result)
```

```python
#20. use replace(), strip(),split(), and join() together

text = input("Enter a sentence: ")

# Remove extra spaces around the string
text = text.strip()

# Replace multiple spaces with a single space
words = text.split()

# Join words with a single space
result = " ".join(words)

# Replace unwanted punctuation
result = result.replace(" ,", ",")
result = result.replace(" .", ".")

# Make first letter uppercase
result = result.capitalize()

print("Cleaned sentence:", result)
```
