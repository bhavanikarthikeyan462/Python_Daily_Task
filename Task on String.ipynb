# Task on String and Notes

```python
# 1. print first 5 characters

text = input("Enter a String:")
print("First 5 Characters:", text[:5])
```

```python
# 2. print last 4 characters

text = input("Enter a string: ")

print("Last 4 characters:", text[-4:])
```

```python
# 3. Characters from index 2 to 8

text = input("Enter a string: ")

print("Characters from index 2 to 8:", text[2:8])
```

```python
# 4. every 2nd character

text = input("Enter a string: ")

print("Every 2nd character:", text[::2])
```

```python
# 5. First name and last name using slicing

name = input("Enter your full name: ")

space = name.find(" ")

first_name = name[:space]
last_name = name[name.rfind(" ") + 1:]

print("First Name:", first_name)
print("Last Name:", last_name)
```

```python
# 6. Uppercase,Lowercase,Title case,captitalize

text = input("Enter a string: ")

print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Title Case:", text.title())
print("Capitalized:", text.capitalize())
```

```python
# 7. Original and Title Case

sentence = input("Enter a sentence: ")

converted = sentence.title()

print("Original:", sentence)
print("Title Case:", converted)
```

```python
# 8. check whether lower() and upper() change original string

text = input("Enter a mixed case string: ")

print("Original:", text)

lower_text = text.lower()
upper_text = text.upper()

print("After lower():", lower_text)
print("After upper():", upper_text)

print("Original after methods:", text)
```

```python
# 9. Check Whether String Contains Only Alphabets

text = input("Enter a string: ")

if text.isalpha():
    print("The string contains only alphabets")
else:
    print("The string does not contain only alphabets")
```

```python
# 10.check if input contains only digits

text = input("Enter a value: ")

if text.isdigit():
    print("It contains only digits")
else:
    print("It does not contain only digits")
```

```python
# 11. check if string is Alphanumeric

text = input("Enter a string: ")

if text.isalnum():
    print("The string is alphanumeric")
else:
    print("The string is not alphanumeric")
```

```python
# 12. check if string is completely lowercase

text = input("Enter a string: ")

if text.islower():
    print("The string is in lowercase")
else:
    print("The string is not in lowercase")
```

```python
# 13. valid password

password = input("Enter password: ")

if password.isalnum() and len(password) >= 8:
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
# 14. find first occurence of a character

text = input("Enter a string: ")
char = input("Enter a character to find: ")

position = text.find(char)

print("Position:", position)
```

```python
# 15. check if a word exists using find()

text = input("Enter a string: ")
word = input("Enter a word to find: ")

position = text.find(word)

if position != -1:
    print("Word found at position:", position)
else:
    print("-1")
```

```python
# 16. find position of python using index()

sentence = input("Enter a sentence:")

position = sentence.index("Python")

print("Position of Python:", position)
```

```python
# 17. diffrence between find() and index()

text = input("Enter a string: ")
char = input("Enter a character that does not exist: ")

print("Using find():", text.find(char))

try:
    print("Using index():", text.index(char))
except ValueError:
    print("Using index(): Character not found")
```

```python
# 18. full name operations

name = input("Enter your full name: ")

print("Uppercase:", name.upper())

space_position = name.find(" ")

print("Position of first space:", space_position)

first_name = name[:space_position]
last_name = name[space_position + 1:]

print("First Name:", first_name)
print("Last Name:", last_name)
```

```python
# 19. Mobile Number validation

mobile = input("Enter mobile number: ")

if len(mobile) == 10 and mobile.isdigit():
    print("Valid Mobile Number")
else:
    print("Invalid Mobile Number")
```

```python
# 20. Reverse, title case and palidrome

text = input("Enter a string: ")

reverse = text[::-1]

print("Original string:", text)
print("Reverse:", reverse)
print("Title Case:", text.title())

if reverse == text:
    print("Palindrome")
else:
    print("Not a Palindrome")
```
