# OOP – Class, Object, Encapsulation (13–18)

```python
# 13. Student class

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)


s1 = Student("Rahul", 85)
s2 = Student("Priya", 92)

s1.display()
s2.display()
```

```python
# 14. BankAccount with private balance

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def display_balance(self):
        print("Balance:", self.__balance)


account = BankAccount(10000)

account.deposit(2000)
account.withdraw(3000)
account.display_balance()
```

```python
# 15. Employee with getter and setter

class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary >= 5000:
            self.__salary = salary
        else:
            print("Salary cannot be below 5000")


emp = Employee(10000)

print("Salary:", emp.get_salary())

emp.set_salary(15000)
print("Updated Salary:", emp.get_salary())

emp.set_salary(3000)
```

```python
# 17. Product with private price

class Product:
    def __init__(self, price):
        self.__price = 0
        self.set_price(price)

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price >= 0:
            self.__price = price
        else:
            print("Price cannot be negative")


p = Product(500)

print("Price:", p.get_price())

p.set_price(800)
print("Updated Price:", p.get_price())

p.set_price(-100)
```

```python
# 18. Rectangle with private length and width

class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def set_length(self, length):
        if length > 0:
            self.__length = length

    def set_width(self, width):
        if width > 0:
            self.__width = width

    def area(self):
        return self.__length * self.__width


r = Rectangle(10, 5)

print("Length:", r.get_length())
print("Width:", r.get_width())
print("Area:", r.area())

r.set_length(20)
r.set_width(10)

print("New Area:", r.area())
```

# Inheritance + Abstraction + super() (19–24)

```python
# 19. Single-level inheritance: Person → Student

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)


s = Student("Rahul", 20, 85)
s.display()
```

```python
# 20. Multi-level inheritance: Person → Employee → Manage

class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary


class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Department:", self.department)


m = Manager("Amit", 50000, "IT")
m.display()
```

```python
# 21. Hierarchical inheritance: Animal → Dog and Cat

class Animal:
    def eat(self):
        print("Animal eats")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.eat()
dog.sound()

cat.eat()
cat.sound()
```

```python
# 22. Multiple inheritance

class Father:
    def father_quality(self):
        print("Father: Hardworking")


class Mother:
    def mother_quality(self):
        print("Mother: Caring")


class Child(Father, Mother):
    def child_quality(self):
        print("Child: Studying")


c = Child()

c.father_quality()
c.mother_quality()
c.child_quality()
```

```python
# 23. Abstract class with area()

from abc import ABC, abstractmethod
import math


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


c = Circle(5)
r = Rectangle(10, 5)

print("Circle Area:", c.area())
print("Rectangle Area:", r.area())
```

```python
# 24. Hybrid inheritance

class A:
    def show_a(self):
        print("Class A")


class B(A):
    def show_b(self):
        print("Class B")


class C(A):
    def show_c(self):
        print("Class C")


class D(B, C):
    def show_d(self):
        print("Class D")


obj = D()

obj.show_a()
obj.show_b()
obj.show_c()
obj.show_d()
```

# Method Overloading, Overriding, Operator Overloading + Duck Typing (25–30)

```python
# 25. Calculator – method overloading concept using *args

class Calculator:
    def add(self, *args):
        return sum(args)


calc = Calculator()

print("Addition of 2 numbers:", calc.add(10, 20))
print("Addition of 3 numbers:", calc.add(10, 20, 30))
```

```python
# 26. Method overriding

class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")


animal = Animal()
dog = Dog()
cat = Cat()

animal.sound()
dog.sound()
cat.sound()
```

```python
# 27. Operator overloading – + for Point

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print("(", self.x, ",", self.y, ")")


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2

p3.display()
```

```python
# 28. Operator overloading – > for Book

class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __gt__(self, other):
        return self.pages > other.pages


book1 = Book("Python Basics", 300)
book2 = Book("Advanced Python", 500)

if book1 > book2:
    print("Book 1 has more pages")
else:
    print("Book 2 has more pages")
```

```python
# 29. Duck Typing

class Duck:
    def walk(self):
        print("Duck is walking")


class Person:
    def walk(self):
        print("Person is walking")


def make_walk(obj):
    obj.walk()


duck = Duck()
person = Person()

make_walk(duck)
make_walk(person)
```

```python
# 30. ComplexNumber with + and - operator overloading

class ComplexNumber:
    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

    def __add__(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imaginary + other.imaginary
        )

    def __sub__(self, other):
        return ComplexNumber(
            self.real - other.real,
            self.imaginary - other.imaginary
        )

    def display(self):
        print(self.real, "+", self.imaginary, "i")


c1 = ComplexNumber(5, 3)
c2 = ComplexNumber(2, 1)

c3 = c1 + c2
c4 = c1 - c2

print("Addition:")
c3.display()

print("Subtraction:")
c4.display()
```
